# PART 1: UNDERSTAING CONTROL STRUCTURES

Control structures determine the order in which instructions in Python program are executed.

Normally, Python executes instructions from top to bottom, one after another. However, sometimes a program needs to:
* Make decisions depending on certain conditions.
* Repeat instructions several times.
* Stop repeating when a particular condition is met.
* Skip certain instructions.
* Process multiple values and calculate results.

This is where **control structures** become important

## Three Main Categories

### 1. Sequential Execution
Instructions execute one after another, in the order they appear.

````python
name = "John Doe"

age = 21

print(name)
print(age)
````

### 2. Selection (Decision-Making)
The program chooses which instructions to execute depending on a condition.

````python
age = 20

if age >= 18:
    print("Eligible")
else:
    print("Not Eligible!")
````

### 3. Repetition (Iteration/Loop)
The program repeats instructions while a condition is true or for each item in a collection.

````python
for number in range(5):
    print(number)
````

**Important distinction**

* Selection determines _which path_ a program follows.
* Iteration determines _how many times_ instructions are executed.


How it works:
1. Python evaluates the condition.
2. If the condition is `True`, the indented instructions execute.
3. If the condition is `False`, Python skips those instructions and continues with the next statement.

# PART 4: The `if`...`else` STATEMENT

The `if...else` statement allows a program to choose between two possible outcomes.

- Unlike a simple `if` statement, which only executes when a condition is `true`, `if ... else` provides an alternative when the condition is `false`.

**Syntax**
```
if condition:
    statement_if_true
else:
    statement_if_false
```

````python
### Basic Implementation
grade = 57

if grade >= 60:
    print("PASSED")
else:
    print("FAILED")
````

Python checks the condition:
* If `grade >= 60` is true, it executes the first block.
Otherwise, it executes the `else` block.

Only one of the two blocks executes.

````python
### Practical example: Checking eligibilty
'''Suppose a personmust be at least 18 years old to register.'''
age = int(input("Enter your age: "))

if age >= 18:
    print("You are eligible to register.")
else:
    print("You are not eligible to register.")
````

##### Here:
* input() collects information from the user. 
* int() converts the entered value into an integer.
* if checks the age.
* else handles the alternatives

# PART 5: THE `if...elif...else` STATEMENT

Sometimes a program needs to evaluate more than two possible outcomes.

For example, a grading system may have five different grades: A, B, C, D and F.

Using several separate if statements is possible, but `if...elif...else is` more appropriate when only one outcome should be selected.

**Understanding the Keywords**

* `if`: Checks the first condition.
* `elif`: Checks another condition if the previous conditions were false.
* `else`: Execute when none of the conditions are true.

***Syntax***
```
if condition1:
    statements
    
elif condition2:
    statements2
    
elif condition3:
    statements3
    
else:
    statements
```

- One can have multiple `elif` statements, but only one `if` statement at the beginning.

##### --- Example --- #####

#### Student Grading System
| Marks | Grade |
|---|---|
| 90–100 | A |
| 80–89 | B |
| 70–79 | C |
| 60–69 | D |
| Below 60 | F |

````python
### Implementation
marks = int(input("Enter your marks: "))

if marks >= 90 and marks <=100:
    grade = "A"
elif 80 <= marks <= 89:
    grade = "B"
elif 70 <= marks <= 79:
    grade = "C"
elif 60 <= marks <= 69:
    grade = "D"
elif 0 <= marks < 60:
    grade = "F"
else:
   grade = "Invalid Grade"

print(grade)
````

**How Python evaluates the conditions**

Suppose the grade is 77.

1. Is 77 greater than or equal to 90? No.
2. Is 77 greater than or equal to 80? No.
3. Is 77 greater than or equal to 70? Yes.
4. Execute print("Grade C").
5. Ignore all remaining conditions.

**Important**

Python evaluates conditions from top to bottom and executes the first matching block.

_Incorrect ordering_:
``` shell
if grade >= 60:
    print("D")
elif grade >= 90:
    print("A")
```

A grade of 95 would produce D because the first condition is already true.

- Always arrange overlapping conditions from the highest threshold to the lowest.

````python
## Practical implementation with input validation

marks = int(input("Enter your marks (0-100): "))

if marks < 0 or marks > 100:
    print("Invalid marks")
    marks = int(input("Enter your marks (0-100): "))

elif 90 <= marks <=100:
    grade= "A"
elif 80<= marks <= 89:
    grade = "B"
elif 70 <= marks <= 79:
    grade = "C"
elif 60 <= marks <= 69:
    grade = "D"
elif 0 <= marks <= 59:
    grade = "F"
else:
    grade = "Invalid grade"
    
print(grade)
````

# PART 6: THE `while` LOOP

A `while` loop repeatedly executes a block of code as long as a specified condition remains `True`.

- It is useful when you do not know in advance how many times the instructions need to execute.

**For example**
- Asking a user to enter a password until it is correct.
- Repeating a menu until the user chooses to exit.
- Processing transactions until there are no more transactions.
- Repeating calculations until a particular result is reached.

***Syntax***
```shell
while condition:
    statement
```

**How it works**
1. Python checks the condition.
2. If it is true, the loop body executes.
3. Python returns to the condition.
4. If the condition remains true, the loop executes again.
5. When the condition becomes false, the loop terminates.

````python
### Basic example

number = 1

while number <= 5:
    print(number) ## This prints out the number, then increments.
    number += 1
    # print(number) ## This increments, then prints out the number 
````

**Understanding the Code Step by Step**

Initially:
```shell
number = 1
```

Python checks:
```shell
number <= 5
```

| Iteration | Number | Condition | Action |
| --- | --- | --- | --- |
| 1 | 1 | True | Print 1 |
| 2 | 2 | True | Print 2 |
| 3 | 3 | True | Print 3 |
| 4 | 4 | True | Print 4 |
| 5 | 5 | True | Print 5 |
| 6 | 6 | False | Stop |

The instruction:

```shell
number += 1 ### number = number + 1
```
increases the value of number by one during every iteration.

````python
### --- Example - Multiplication by 3 --- ###

product = 1

while product <= 50:
    print(product)
    product *= 3
````

````python
product = 1

while product <= 50:
    product = product * 3

print(product)
````

| Iteration | Calculation | Product |
| --- | --- | --- |
| Initial | Starting value | 1 |
| 1 | 1 × 3 | 3 |
| 2 | 3 × 3 | 9 |
| 3 | 9 × 3 | 27 |
| 4 | 27 × 3 | 81 |

When the product reaches 81, the condition product <= 50 becomes false, so the loop stops.

##### Infinite loops
An infinite loop occurs when a loop condition never becomes false.

```shell
number = 1

while number <= 5:
    print(number)
```
The problem is that number is never increased.
- Therefore, the condition always remains true.

To correct it:

```shell
number = 1

while number <= 5
    print(number)
    number = number + 1
```

**Examination Tip**

Whenever you use a `while` loop, identify:
- The initial value.
- The condition.
- The update statement.

These three components help prevent infinite loops.

## The `range()` Function

The `range()` function generates a sequence of integers, commonly used with for loops.
- It is particularly useful when you want to repeat an ioperation a specific number of times.

There are three (3) ways to use it:

#### One argument: `range(stop)`

````python
for number in range(5):
    print(number)
````

The sequence starts at 0 and ends at 4.

**Important**: The stop value is excluded.

Therefore:
```bash
range(5)
```
generates five numbers, not six.

#### Two arguments: `range(start, stop)`

You can specify where the sequence should begin and where it should stop.

````python
for number in range (5, 10):
    print(number)
````

The `first argument` is the `starting value`, while the `second argument` is the `stopping value` (_excluded_).

#### Three arguments: range(start, stop, step)

The `third` argument determines the incremeny or decrement.

````python
for numbers in range(5, 50, 5):
    print(numbers)
````

````python
for numbers in range(0, 10, 2):
    print(numbers)
````

The sequence increases by 2 each time.

**Counting backwards**

One can use a negative step to generate a descending sequence.

````python
for numbers in range(10, -2, -2):
    print(numbers)
````

**Understanding the three forms**

| Function | Generated sequence |
| --- | --- |

| `range(0, 10, 2)` | 0, 2, 4, 6, 8 |
| `range(10, 0, -2)` | 10, 8, 6, 4, 2 |

## `break` and `continue` Statements

Both statements change how a loop executes.

However, they perform different functions.

#### The `break` statement

The `break` statement immediately terminates the entire loop.

````python
for number in range(5, 1000, 5):
    if number == 50:
        break

    print(number)
````

`range(5, 1000, 5)`:
* Start (5): The sequence begins at 5.
* Stop (1000): The sequence goes up to, but does not include, 1000 (it stops at 995).
* Step (5): It increases by 5 after each loop (5, 10, 15, 20...)

`if number == 50: break`:
* During each loop, Python checks if the current number equals 50.
* The moment number reaches 50, the break statement immediately exits and terminates the loop.
* Because break happens before the print statement for 50, the number 50 is never printed

`print(number)`: 
* As long as the break condition is false (for numbers less than 50), it prints the current value of number on a new line

````python
for number in range(0,20,2):
    if number == 18:
        break
    print(number)
````

````python
for numbers in range(0, 25, 2):
    if numbers >= 17:
        print("The number is 17")
        break
    print(numbers)
````

#### The `continue` Statement

The `continue` statement skips the remaining instructions in the current iteration and moves to the next iteration.

````python
for number in range(5, 50, 5):
    if number == 10:
        continue
    print(number)
````

Notice that the loop does not terminate. It `simply skips 10`.

**Difference between break and continue**

````python
# break
# Terminates the loop completely.
for i in range(5):
    if i == 2:
        break
    print(i)
````

````python
# continue
# Skips only the current iteration.
for i in range(5):
    if i == 2:
        continue
    print(i)
````

````python
# Practical example: Finding a number
numbers = [10, 20, 50, 40, 90, 30]

for number in numbers:
    if number == 40:
        print("Number found")
        break

    print(number)
````

````python
# Practical example: Skipping even numbers

for numbers in range(0,10):
    if numbers % 2 == 0:
        continue
        
    print(numbers)
````

## Understaing the Problem

````python
'''
A class of ten students takes a quiz. Their grades are:
'''
[98, 76, 71, 87, 83, 90, 57, 79, 82, 94]
````

##### Develop a program that calculates the class average
An algorithm is a step-by-step procedure for solving a problem.

To calculate the average:
$$
Average = \frac{\text{Sum of all grades}}{\text{Number of grades}}
$$

**We therefore need to**:
1. Store the grades.
2. Initialize a variable to store the total.
3. Initialize a counter to count the grades.
4. Loop through the grades.
5. Add each grade to the total.
6. Increase the counter for every grade.
7. Calculate the average.
8. Display the result.

````python
# store the grades
class_grades = [98, 76, 71, 87, 83, 90, 57, 79, 82, 94]

# Initialize variables
class_total = 0

# Using a for loop 
for grades in class_grades:
    class_total += grades
    class_avg = class_total / len(class_grades)

# Display result
print(f"Class average: {class_avg}")
````

````python
### Alternative approach

# Store variables 
class_grades = [98, 76, 71, 87, 83, 90, 57, 79, 82, 94]

# Initialize variables 
class_total = 0

grade_counter = 0

# Calculate total and count grades
for grade in class_grades:
    class_total += grade
    grade_counter +=1

# Calculate average 
class_average = class_total / grade_counter

# Display result
print(f"Class average is: {class_average:.2f}")
````

**Understading the important variables**
| Variable | Purpose |
| --- | --- |

| `total` | Stores the running sum |

# PART 9: SENTINEL-CONTROLLED ITERATION

Instead of processing a predefined list, the program accepts grades from the user until the user enter a special value.
- That special value is called a **sentinel value**,
- The difference is that the number of grades is no longer known in advance.

**What is sentinel-controlled iteration?**

Sentinel-controlled iteration is a repetition technique where a loop continues until the user enters a particular value that signals the end of input.

For example, a program can accept an unlimited number of grades until the user enters `-1`.

_The sentinel value_:
* Controls when the loop stops.
* Is not treated as valid data.
* Must be distinguishable from valid inputs.

**Understanding the algorithm**

Requirements:
Develop a class-average program that accepts an arbitrary number of grades.
1. Initialize `total` to 0.
2. Initialize `count` to 0.
3. Ask the user to enter a grade.
4. Check whether the grade is equal to -1.
5. If it is not -1, add it to the total.
6. Increase the counter.
7. Request another grade.
8. Repeat until -1 is entered.
9. Calculate and display the average.
10. If no grades were entered, display an appropriate message.

````python
# Initialise an empty variables
total_grades = 0
counter = 0

# Ask for the first grade
class_grades = int(input("Enter grade: (or -1: to exit): "))

#### Test
# print(class_grades)

while class_grades != -1:
    # compute the total
    total_grades += class_grades
    # total_grades.append(class_grades)

    # Keep count increment by 1
    counter += 1

    # Ask for the grade or exit the loop
    class_grades = int(input("Enter grade: (or -1: to exit): "))

    #### Test 
    # print(counter)

# Calculate and display the class average
if counter > 0:
    class_average = total_grades / counter
    print(f"The class average: {class_average:.2f}")
else:
    print("No grades were entered")
````

# PART 10: NESETED CONTROL STATEMENTS

A nested control statement is a control structure palced inside another control structure.

For example:
* An `if` statement inside another `if` statement.
* An `if` statement inside a `for` loop.
* A `for` loop inside a `while` loop.
* A `while` loop inside an `if` statement.

Nesting allows programs to handle more complex situations.

**Simple example: Nested if statements**

Suppose a university offers a scholarship to students who:
- Have scored at least 80 marks.
- Have attended at least 90% of their classes.

````python
grade = 80
attendance = 95

if grade >= 80:

    if attendance >= 90:
        print("Scholarship awarded.")
    else:
        print("Attendance requirement not met.")
else:
    print("Grade requirement not met.")
````

**Explanation**:
1. Python first checks whether the grade is at least 80.
2. If true, it checks the attendance.
3. If both conditions are met, the scholarship is awarded.
4. Otherwise, the relevant message is displayed.

**Nested Loops**

A loop can also contain another looop 

````python
for i in range(1,4):
    for j in range(1,4):
        print(i,j)
````

The outer loop executes three tiimes.
- For every iteration of the outer loop, the inner loop executes three times.

Therefor: 
$3 * 3 = 9 iterations$

````python
# Multiplication Table

for i in range(1,4):
    for j in range(1,4):
        print(f"{i*j}", end='\t')
    print()
````

* What happens here: Each print() line builds one horizontal row of the table.
* How `sep='\t'` works: By default, `print()` puts a single space between items separated by commas.
* Setting `sep='\t'` swaps that space for a tab.

````python
print(1 * 1, 1 * 2, 1 * 3, sep='\t')
print(2 * 1, 2 * 2, 2 * 3, sep='\t')
print(3 * 1, 3 * 2, 3 * 3, sep='\t')
````

**Spot the Pattern**
Look at the three lines above. Notice that the first number in every multiplication changes from 1 to 2 to 3, while everything else stays identical:

* Row 1: 1 * 1, 1 * 2, 1 * 3

* Row 2: 2 * 1, 2 * 2, 2 * 3

* Row 3: 3 * 1, 3 * 2, 3 * 3


**Replace the Repeating Number with i**

Because only that first number changes, you can replace it with a variable named i and let a for loop count from 1 to 3 for you:

````python
for i in range(1,4):
    print(i * 1, i * 2, i * 3, sep='\t')
````

* Using [1, 2, 3] instead of range(1, 4): Writing a plain list [1, 2, 3] shows the exact numbers i will become, without needing to remember that range(1, 4) stops before 4.
* First pass (i = 1): Runs print(1 * 1, 1 * 2, 1 * 3, sep='\t') and outputs 1	2	3.
* Second pass (i = 2): Runs print(2 * 1, 2 * 2, 2 * 3, sep='\t') and outputs 2	4	6.
* Third pass (i = 3): Runs print(3 * 1, 3 * 2, 3 * 3, sep='\t') and outputs 3	6	9.


## Boolean Operators

Boolean operators allow you to combine multiple conditions into one expression.

They are useful when a decision depends on more than one requirement.

Python has three main Boolean operators:
1. `and`
2. `or`
3. `not`

### The `and` operator
The and operator returns True only when all conditions are true.

````python
gender = "female"

age = 78

if gender == "female" and age == 78:
    print('Senior female')
````

Both conditions must be satisfied.

Truth table:

| Condition A | Condition B | A and B |
| --- | --- | --- |
| True | True | True |
| True | False | False |
| False | True | False |
| False | False | False |

````python
## **Practical example: University admission.**

marks = 75
attendance = 85

if marks >= 75 and attendance >= 80:
    print("Admission requirements met")
else:
    print("Admission requirements not met")
````

### The `or` operator
The or operator returns True when at least one condition is true.

````python
semester_average = 83
final_exam = 95

if semester_average >= 90 or final_exam >= 90:
    print("Student gets an A")
````

Only one condition needs to be true.

Truth table:

| Condition A | Condition B | A or B |
| --- | --- | --- |
| True | True | True |
| True | False | True |
| False | True | True |
| False | False | False |

````python
### Practical example: Accessing a student portal.

### A user can log in using either their student ID or registered email, provided the credentials are valid.

student_id = False
registered_email = True 

if student_id or registered_email:
    print("Login identifier accepted")
````

### The `not` operator
The not operator reverses a Boolean value.

- not True becomes False.

- not False becomes True.

````python
game_over = False 

if not game_over:
    print("Continue playing")
````

Truth table:

| Condition | Not condition |
| --- | --- |
| True | False |
| False | True |

You can also use not to check whether something is not equal to a particular value.

````python
grade = 85

if not grade == -1:
    print("Valid grade")
````

````python
## Alternative 

if grade != -1:
    print("Valid grade")
````

### Combining Boolean operators
You can combine and, or and not in one condition.

Example:

A student qualifies for an award if they:
- Have at least 85 marks and at least 90% attendance; OR
- Have achieved at least 95 marks.

````python
### Initialize the variables 

marks = 85
attendance = 90 

# 1. Validate inputs first
if (marks > 100 or marks < 0):
    print("Invalid Marks Input")
elif (attendance < 0 or attendance > 100):
    print("Invalid Attendance value")

# 2. Check award eligibility only if inputs are valid

elif (marks >= 85 and attendance >= 90) or  marks >= 95:
    print("Student qualifies for an award") 
else:
    print("Student does not qualify for an award")
````

# PART 11: MEAN, MEDIAN and MODE

These are statistical measures used to summarize a collection of numerical values.

Python provides the statistics module, which includes functions for calculating these measures.

## Mean
The mean is the arithmetic average of a collection of numbers.

Formula:
$$
Mean = \frac{\text{Sum of all values}}{\text{Number of values}}
$$

````python
grades = [30, 10, 25, 5, 30]

mean = sum(grades) / len(grades)

print(f"Mean: {mean:.2f}")
````

Calculation:
$$
Mean = \frac{30+10+25+5+30}{5}
$$

$$
Mean = 20
$$

````python
### Python Implementation 

import statistics

grades = [30, 10, 25, 5, 30]

print(f"Mean: {statistics.mean(grades):.2f}")
````

## Median
The median is the middle value when numbers are arranged in ascending or descending order.

**Important**: _Always arrange the values in order before identifying the median._

Example with an odd number of values:

````python
grades = [30, 10, 25, 5, 30]

print(sorted(grades))
````

The middle value is 25.

Therefore:
$$
Median = 25
$$

````python
import statistics 

grades = [30, 10, 25, 5, 30]

print(f"Median value: {statistics.median(grades)}")
````

##### What happens when there is an even number of values?

The median is the average of the two middle values.

````python
numbers = [10, 20, 30, 40]
````

The two middle values are 20 and 30.
$$
Median = \frac{20+30}{2}=25
$$

````python
numbers = [10, 20, 30, 40]

print(f"Median Value {statistics.median(numbers):.2f}")
````

## Mode
The mode is the value that occurs most frequently in a dataset.

Example:

````python
grades = [30, 10, 25, 5, 30]
````

**Frequency:**

| Grade | Frequency |
| --- | --- |
| 30 | 2 |
| 10 | 1 |
| 25 | 1 |
| 5 | 1 |

Therefore:
$$
Mode = 30
$$

````python
print(f"Mode Value: {statistics.mode(grades)}")
````

**Comparing the three measures**

| Measure | What it tells us | How to calculate |
| --- | --- | --- |
| Mean | Average value | Sum ÷ count |
| Median | Middle value | Arrange and find middle |
| Mode | Most frequent value | Find most repeated |

**Additional example**

Consider the following test scores:

````python
scores = [60, 70, 70, 80, 90, 90, 90]

print(f"Mean Value: {statistics.mean(scores):.2f}")
print(f"Median Value: {statistics.median(scores)}")
print(f"Mode Value: {statistics.mode(scores)}")
````

**Understanding when to use each measure**:

- Mean is useful when you want the overall average.

- Median is useful when extreme values may distort the average.

- Mode is useful when identifying the most common value.

For example, a business may use the mean to analyse average sales, the median to understand typical transaction values, and the mode to identify its most frequently purchased product.
