# PART 1: UNDERSTAING CONTROL STRUCTURES

Control structures determine the order in which instructions in Python program are executed.

Normally, Python executes instructions from top to bottom, one after another. However, sometimes a program needs to:
* Make decisions depending on certain conditions.
* Repeat instructions several times.
* Stop repeating when a particular condition is met.
* Skip certain instructions.
* Process multiple values and calculate results.

This is where **control structures** become important

## Three Main Categories

### 1. Sequential Execution
Instructions execute one after another, in the order they appear.

````python
name = "John Doe"

age = 21

print(name)
print(age)
````

### 2. Selection (Decision-Making)
The program chooses which instructions to execute depending on a condition.

````python
age = 20

if age >= 18:
    print("Eligible")
else:
    print("Not Eligible!")
````

### 3. Repetition (Iteration/Loop)
The program repeats instructions while a condition is true or for each item in a collection.

````python
for number in range(5):
    print(number)
````

**Important distinction**

* Selection determines _which path_ a program follows.
* Iteration determines _how many times_ instructions are executed.

# PART 3: THE `if` STATEMENT

The `if` statement allows a program to execute an instruction only when a specified condition is true

Think of it as asking a question:

_If a student scores at least 60, display "PASSED"_

**Syntax**
```
if condition:
    statement
```

How it works:
1. Python evaluates the condition.
2. If the condition is `True`, the indented instructions execute.
3. If the condition is `False`, Python skips those instructions and continues with the next statement.

##### Basic Implementation 

````python
grade = 80

if grade >= 60:
    print("PASSED")
````

````python
'''Changing the Grade and see what happens'''

grade = 45

if grade >= 60:
    print("PASSED")
````

There is no output because the condition is `False`.

#### Understanding Indentation 
Indentation refers to the spaces at the beginning of a line.

Python uses indentation to identify which instructions belong to a particular control structure.

````python
grade = 61

if grade >= 60:
    print("PASSED")
    print("Good Job!")
````

````python
###### ---- Incorrect ---- ######

if grade >= 60:
print("PASSED")  # IndentationError: expected an indented block after 'if' statement on line 3
````

The second example produces an IndentationError.

#### Important Rules:
* Always use a colon `:` after an `if` condition.
* Indent the instructions inside the `if` block.
* Use consistent indentaion, normally four spaces || (tab).
* Instructions outside the block should return to their original indentation level.

# PART 4: The `if`...`else` STATEMENT

The `if...else` statement allows a program to choose between two possible outcomes.

- Unlike a simple `if` statement, which only executes when a condition is `true`, `if ... else` provides an alternative when the condition is `false`.

**Syntax**
```
if condition:
    statement_if_true
else:
    statement_if_false
```

````python
### Basic Implementation
grade = 57

if grade >= 60:
    print("PASSED")
else:
    print("FAILED")
````

Python checks the condition:
* If `grade >= 60` is true, it executes the first block.
Otherwise, it executes the `else` block.

Only one of the two blocks executes.

````python
### Practical example: Checking eligibilty
'''Suppose a personmust be at least 18 years old to register.'''
age = int(input("Enter your age: "))

if age >= 18:
    print("You are eligible to register.")
else:
    print("You are not eligible to register.")
````

##### Here:
* input() collects information from the user. 
* int() converts the entered value into an integer.
* if checks the age.
* else handles the alternatives

**Assigning Values Using Conditions**

- One can also assign different values to a variable depending on the condition.

````python
grade = 87

if grade >= 60:
    result = "Passed"
else:
    result = "Failed"

print(result)
````

**Ternary Conditional Expression**

- Python provides a shorter way of writing a simple `if ... else` statement.

***Syntax***
```
variable = value_if_true if condition else value_if_false
```

````python
results = "Passed" if grade >= 60 else "Failed"
````

````python
print(results)
````

This produces the same result as the longer `if...else` implementation.

Use _ternary expressions_ for simple decisions. 
- For complex conditions, the `normal if...else structure` is easier to understand.

# PART 5: THE `if...elif...else` STATEMENT

Sometimes a program needs to evaluate more than two possible outcomes.

For example, a grading system may have five different grades: A, B, C, D and F.

Using several separate if statements is possible, but `if...elif...else is` more appropriate when only one outcome should be selected.

**Understanding the Keywords**

* `if`: Checks the first condition.
* `elif`: Checks another condition if the previous conditions were false.
* `else`: Execute when none of the conditions are true.

***Syntax***
```
if condition1:
    statements
    
elif condition2:
    statements2
    
elif condition3:
    statements3
    
else:
    statements
```

- One can have multiple `elif` statements, but only one `if` statement at the beginning.

##### --- Example --- #####

#### Student Grading System
| Marks | Grade |
|---|---|
| 90–100 | A |
| 80–89 | B |
| 70–79 | C |
| 60–69 | D |
| Below 60 | F |

````python
### Implementation
marks = int(input("Enter your marks: "))

if marks >= 90 and marks <=100:
    grade = "A"
elif 80 <= marks <= 89:
    grade = "B"
elif 70 <= marks <= 79:
    grade = "C"
elif 60 <= marks <= 69:
    grade = "D"
elif 0 <= marks < 60:
    grade = "F"
else:
   grade = "Invalid Grade"

print(grade)
````

**How Python evaluates the conditions**

Suppose the grade is 77.

1. Is 77 greater than or equal to 90? No.
2. Is 77 greater than or equal to 80? No.
3. Is 77 greater than or equal to 70? Yes.
4. Execute print("Grade C").
5. Ignore all remaining conditions.

**Important**

Python evaluates conditions from top to bottom and executes the first matching block.

_Incorrect ordering_:
``` shell
if grade >= 60:
    print("D")
elif grade >= 90:
    print("A")
```

A grade of 95 would produce D because the first condition is already true.

- Always arrange overlapping conditions from the highest threshold to the lowest.

````python
## Practical implementation with input validation

marks = int(input("Enter your marks (0-100): "))

if marks < 0 or marks > 100:
    print("Invalid marks")
    marks = int(input("Enter your marks (0-100): "))

elif 90 <= marks <=100:
    grade= "A"
elif 80<= marks <= 89:
    grade = "B"
elif 70 <= marks <= 79:
    grade = "C"
elif 60 <= marks <= 69:
    grade = "D"
elif 0 <= marks <= 59:
    grade = "F"
else:
    grade = "Invalid grade"
    
print(grade)
````

# PART 6: THE `while` LOOP

A `while` loop repeatedly executes a block of code as long as a specified condition remains `True`.

- It is useful when you do not know in advance how many times the instructions need to execute.

**For example**
- Asking a user to enter a password until it is correct.
- Repeating a menu until the user chooses to exit.
- Processing transactions until there are no more transactions.
- Repeating calculations until a particular result is reached.

***Syntax***
```shell
while condition:
    statement
```

**How it works**
1. Python checks the condition.
2. If it is true, the loop body executes.
3. Python returns to the condition.
4. If the condition remains true, the loop executes again.
5. When the condition becomes false, the loop terminates.

````python
### Basic example

number = 1

while number <= 5:
    print(number) ## This prints out the number, then increments.
    number += 1
    # print(number) ## This increments, then prints out the number 
````

**Understanding the Code Step by Step**

Initially:
```shell
number = 1
```

Python checks:
```shell
number <= 5
```

| Iteration | Number | Condition | Action |
| --- | --- | --- | --- |
| 1 | 1 | True | Print 1 |
| 2 | 2 | True | Print 2 |
| 3 | 3 | True | Print 3 |
| 4 | 4 | True | Print 4 |
| 5 | 5 | True | Print 5 |
| 6 | 6 | False | Stop |

The instruction:

```shell
number += 1 ### number = number + 1
```
increases the value of number by one during every iteration.

````python
### --- Example - Multiplication by 3 --- ###

product = 1

while product <= 50:
    print(product)
    product *= 3
````

````python
product = 1

while product <= 50:
    product = product * 3

print(product)
````

| Iteration | Calculation | Product |
| --- | --- | --- |
| Initial | Starting value | 1 |
| 1 | 1 × 3 | 3 |
| 2 | 3 × 3 | 9 |
| 3 | 9 × 3 | 27 |
| 4 | 27 × 3 | 81 |

When the product reaches 81, the condition product <= 50 becomes false, so the loop stops.

##### Infinite loops
An infinite loop occurs when a loop condition never becomes false.

```shell
number = 1

while number <= 5:
    print(number)
```
The problem is that number is never increased.
- Therefore, the condition always remains true.

To correct it:

```shell
number = 1

while number <= 5
    print(number)
    number = number + 1
```

**Examination Tip**

Whenever you use a `while` loop, identify:
- The initial value.
- The condition.
- The update statement.

These three components help prevent infinite loops.

Examples of iterables include:
* Strings.
* Lists
* Tuplse
* Ranges
* Dictionaries.

***Syntax***

The variable represents the current item being processed.

**Looping through a string**

A string is a sequence of individual characters.

````python
for characters in "Python":
    print(characters.upper())
````

## The `range()` Function

The `range()` function generates a sequence of integers, commonly used with for loops.
- It is particularly useful when you want to repeat an ioperation a specific number of times.

There are three (3) ways to use it:

#### One argument: `range(stop)`

````python
for number in range(5):
    print(number)
````

The sequence starts at 0 and ends at 4.

**Important**: The stop value is excluded.

Therefore:
```bash
range(5)
```
generates five numbers, not six.

#### Two arguments: `range(start, stop)`

You can specify where the sequence should begin and where it should stop.

````python
for number in range (5, 10):
    print(number)
````

The `first argument` is the `starting value`, while the `second argument` is the `stopping value` (_excluded_).

#### Three arguments: range(start, stop, step)

The `third` argument determines the incremeny or decrement.

````python
for numbers in range(5, 50, 5):
    print(numbers)
````

````python
for numbers in range(0, 10, 2):
    print(numbers)
````

The sequence increases by 2 each time.

**Counting backwards**

One can use a negative step to generate a descending sequence.

````python
for numbers in range(10, -2, -2):
    print(numbers)
````

**Understanding the three forms**

| Function | Generated sequence |
| --- | --- |
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(2, 7)` | 2, 3, 4, 5, 6 |
| `range(0, 10, 2)` | 0, 2, 4, 6, 8 |
| `range(10, 0, -2)` | 10, 8, 6, 4, 2 |

````python
### Practical Implementation: Multiplication Table

number = int(input("Enter a number: "))

for i in range(1,11):
    total = number * i
    print(f"{number} * {i} = {total}")
    
````

**Understanding off-by-one errors**

An off-by-one error occurs when a loop executes one fewer or one more time than intended.

For example:
```bash
for number in range(1, 10):
    print(number)
```
This generates `1 to 9`, not `1 to 10`.

To include 10:
```bash 
for number in range(1, 11):
    print(number)
```

Always remember that `range()` excludes its _stopping value_.

## `break` and `continue` Statements

Both statements change how a loop executes.

However, they perform different functions.

#### The `break` statement

The `break` statement immediately terminates the entire loop.

````python
for number in range(5, 1000, 5):
    if number == 50:
        break

    print(number)
````

`range(5, 1000, 5)`:
* Start (5): The sequence begins at 5.
* Stop (1000): The sequence goes up to, but does not include, 1000 (it stops at 995).
* Step (5): It increases by 5 after each loop (5, 10, 15, 20...)

`if number == 50: break`:
* During each loop, Python checks if the current number equals 50.
* The moment number reaches 50, the break statement immediately exits and terminates the loop.
* Because break happens before the print statement for 50, the number 50 is never printed

`print(number)`: 
* As long as the break condition is false (for numbers less than 50), it prints the current value of number on a new line

````python
for number in range(0,20,2):
    if number == 18:
        break
    print(number)
````

````python
for numbers in range(0, 25, 2):
    if numbers >= 17:
        print("The number is 17")
        break
    print(numbers)
````

#### The `continue` Statement

The `continue` statement skips the remaining instructions in the current iteration and moves to the next iteration.

````python
for number in range(5, 50, 5):
    if number == 10:
        continue
    print(number)
````

Notice that the loop does not terminate. It `simply skips 10`.

**Difference between break and continue**

````python
# break
# Terminates the loop completely.
for i in range(5):
    if i == 2:
        break
    print(i)
````

````python
# continue
# Skips only the current iteration.
for i in range(5):
    if i == 2:
        continue
    print(i)
````

````python
# Practical example: Finding a number
numbers = [10, 20, 50, 40, 90, 30]

for number in numbers:
    if number == 40:
        print("Number found")
        break

    print(number)
````

````python
# Practical example: Skipping even numbers

for numbers in range(0,10):
    if numbers % 2 == 0:
        continue
        
    print(numbers)
````

The `%` operator calculates the remainder. A number divisible by _2 has a remainder of zero_.

## PART 8: PROGRAM DEVELOPMENT: SEQUENCE-CONTROLLED ITERATION

This section introduces an important programming skill: developing a complete solution to a problem rather than simply writing individual Python statements.

The notebook uses a class-average calculation as its main example.

### Understaing the Problem

````python
'''
A class of ten students takes a quiz. Their grades are:
'''
[98, 76, 71, 87, 83, 90, 57, 79, 82, 94]
````

##### Develop a program that calculates the class average
An algorithm is a step-by-step procedure for solving a problem.

To calculate the average:
$$
Average = \frac{\text{Sum of all grades}}{\text{Number of grades}}
$$

**We therefore need to**:
1. Store the grades.
2. Initialize a variable to store the total.
3. Initialize a counter to count the grades.
4. Loop through the grades.
5. Add each grade to the total.
6. Increase the counter for every grade.
7. Calculate the average.
8. Display the result.

````python
# store the grades
class_grades = [98, 76, 71, 87, 83, 90, 57, 79, 82, 94]

# Initialize variables
class_total = 0

# Using a for loop 
for grades in class_grades:
    class_total += grades
    class_avg = class_total / len(class_grades)

# Display result
print(f"Class average: {class_avg}")
````

````python
### Alternative approach

# Store variables 
class_grades = [98, 76, 71, 87, 83, 90, 57, 79, 82, 94]

# Initialize variables 
class_total = 0

grade_counter = 0

# Calculate total and count grades
for grade in class_grades:
    class_total += grade
    grade_counter +=1

# Calculate average 
class_average = class_total / grade_counter

# Display result
print(f"Class average is: {class_average:.2f}")
````

**Understading the important variables**
| Variable | Purpose |
| --- | --- |
| `grades` | Stores all the students' grades |
| `total` | Stores the running sum |
| `grade_counter` | Counts the number of grades |
| `grade` | Represents the current grade |
| `average` | Stores the calculated average |
