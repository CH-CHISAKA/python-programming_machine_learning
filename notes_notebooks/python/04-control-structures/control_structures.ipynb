# PART 1: UNDERSTAING CONTROL STRUCTURES

Control structures determine the order in which instructions in Python program are executed.

Normally, Python executes instructions from top to bottom, one after another. However, sometimes a program needs to:
* Make decisions depending on certain conditions.
* Repeat instructions several times.
* Stop repeating when a particular condition is met.
* Skip certain instructions.
* Process multiple values and calculate results.

This is where **control structures** become important

## Three Main Categories

### 1. Sequential Execution
Instructions execute one after another, in the order they appear.

````python
name = "John Doe"

age = 21

print(name)
print(age)
````

### 2. Selection (Decision-Making)
The program chooses which instructions to execute depending on a condition.

````python
age = 20

if age >= 18:
    print("Eligible")
else:
    print("Not Eligible!")
````

### 3. Repetition (Iteration/Loop)
The program repeats instructions while a condition is true or for each item in a collection.

````python
for number in range(5):
    print(number)
````

**Important distinction**

* Selection determines _which path_ a program follows.
* Iteration determines _how many times_ instructions are executed.

# PART 2: CONDITIONS AND COMPARISON OPERATORS

###### What is a condition?
A condition is an expression that evaluates to one of two Boolean values:
* `True` - the condition is correct.
* `False` - the condition is incorrect.

Conditions are commonly used in `if`, `elif`, `else`, `while` and other control statements

````python
'''Is 10 greater than 5'''
print(10 > 5) # True

'''Is 10 less than 5'''
print(10<5) # False

'''Is 10 equal to 10'''
print(10 == 10) # True
````

## Comparison Operators
These compare two values and return either `True` or `False`

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `>` | Greater than | `10 > 5` | True |
| `<` | Less than | `10 < 5` | False |
| `>=` | Greater than or equal to | `10 >= 10` | True |
| `<=` | Less than or equal to | `5 <= 10` | True |
| `==` | Equal to | `5 == 5` | True |
| `!=` | Not equal to | `5 != 3` | True |

**Difference between `=` and `==`
This is one of the most important distinctions in Python.

**Assignment Operator** `(=)`
- Assigns a value to a variable.

````python
age = 21 ## This means that the variable `age` stores the value `21`.

age
````

**Equality Operator** `(==)`

- Checks whether two values are equal.

````python
age == 21 # True - This checks whether the value in age is 21
````

**Remember**

- One equals sign = means assign.
- Two equals signs == mean compare.

Writing if age = 21: will cause a syntax error.

# PART 3: THE `if` STATEMENT

The `if` statement allows a program to execute an instruction only when a specified condition is true

Think of it as asking a question:

_If a student scores at least 60, display "PASSED"_

**Syntax**
```
if condition:
    statement
```

How it works:
1. Python evaluates the condition.
2. If the condition is `True`, the indented instructions execute.
3. If the condition is `False`, Python skips those instructions and continues with the next statement.

##### Basic Implementation 

````python
grade = 80

if grade >= 60:
    print("PASSED")
````

````python
'''Changing the Grade and see what happens'''

grade = 45

if grade >= 60:
    print("PASSED")
````

There is no output because the condition is `False`.

#### Understanding Indentation 
Indentation refers to the spaces at the beginning of a line.

Python uses indentation to identify which instructions belong to a particular control structure.

````python
grade = 61

if grade >= 60:
    print("PASSED")
    print("Good Job!")
````

````python
###### ---- Incorrect ---- ######

if grade >= 60:
print("PASSED")  # IndentationError: expected an indented block after 'if' statement on line 3
````

The second example produces an IndentationError.

#### Important Rules:
* Always use a colon `:` after an `if` condition.
* Indent the instructions inside the `if` block.
* Use consistent indentaion, normally four spaces || (tab).
* Instructions outside the block should return to their original indentation level.

##### Truthy and falsy values

Python can evaluate values other than explicit Boolean values as true or false.

**Examples of falsy values**:
- 0
- 0.0
- '' (empty string)
- [x] (empty list)
- [ ] (empty list)
- None
- False

**Examples of truthy values**:
- 1, -5, 10
- Non-empty strings.
- Non-empty lists.
- True

````python
##### -- Implementation --#####
name = "Jane Doe"

if name:
    print("Name has been entered")
````

This is useful when checking a user has entered information.

# PART 4: The `if`...`else` STATEMENT

The `if...else` statement allows a program to choose between two possible outcomes.

- Unlike a simple `if` statement, which only executes when a condition is `true`, `if ... else` provides an alternative when the condition is `false`.

**Syntax**
```
if condition:
    statement_if_true
else:
    statement_if_false
```

````python
### Basic Implementation
grade = 57

if grade >= 60:
    print("PASSED")
else:
    print("FAILED")
````

Python checks the condition:
* If `grade >= 60` is true, it executes the first block.
Otherwise, it executes the `else` block.

Only one of the two blocks executes.

````python
### Practical example: Checking eligibilty
'''Suppose a personmust be at least 18 years old to register.'''
age = int(input("Enter your age: "))

if age >= 18:
    print("You are eligible to register.")
else:
    print("You are not eligible to register.")
````

##### Here:
* input() collects information from the user. 
* int() converts the entered value into an integer.
* if checks the age.
* else handles the alternatives

**Assigning Values Using Conditions**

- One can also assign different values to a variable depending on the condition.

````python
grade = 87

if grade >= 60:
    result = "Passed"
else:
    result = "Failed"

print(result)
````

**Ternary Conditional Expression**

- Python provides a shorter way of writing a simple `if ... else` statement.

***Syntax***
```
variable = value_if_true if condition else value_if_false
```

````python
results = "Passed" if grade >= 60 else "Failed"
````

````python
print(results)
````

This produces the same result as the longer `if...else` implementation.

Use _ternary expressions_ for simple decisions. 
- For complex conditions, the `normal if...else structure` is easier to understand.

# PART 5: THE `if...elif...else` STATEMENT

Sometimes a program needs to evaluate more than two possible outcomes.

For example, a grading system may have five different grades: A, B, C, D and F.

Using several separate if statements is possible, but `if...elif...else is` more appropriate when only one outcome should be selected.

**Understanding the Keywords**

* `if`: Checks the first condition.
* `elif`: Checks another condition if the previous conditions were false.
* `else`: Execute when none of the conditions are true.

***Syntax***
```
if condition1:
    statements
    
elif condition2:
    statements2
    
elif condition3:
    statements3
    
else:
    statements
```

- One can have multiple `elif` statements, but only one `if` statement at the beginning.

##### --- Example --- #####

#### Student Grading System
| Marks | Grade |
|---|---|
| 90–100 | A |
| 80–89 | B |
| 70–79 | C |
| 60–69 | D |
| Below 60 | F |

````python
### Implementation
marks = int(input("Enter your marks: "))

if marks >= 90 and marks <=100:
    grade = "A"
elif 80 <= marks <= 89:
    grade = "B"
elif 70 <= marks <= 79:
    grade = "C"
elif 60 <= marks <= 69:
    grade = "D"
elif 0 <= marks < 60:
    grade = "F"
else:
   grade = "Invalid Grade"

print(grade)
````
