The notebook presents a program that calculates how many students passed and failed a licensing examination.

Requirements:
- Process examination results for 10 students.
- A result of 1 means pass.
- A result of 2 means fail.
- Count the number of passes and failures.
- If more than 8 students pass, display a bonus message.

```python
"""
Requesting input for pass-1 or fail-2
"""

num_student = int(input("Enter 1:for pass, 2: for fail: or (0: to exit): "))

students_score = []

while num_student !=0:
    # Check if input is neither 1 nor 2
    
    ### Alternative 1
    # if num_student !=1 and num_student !=2:
    #     print("Invalid input", num_student)
    # else:
    #     # Only append if the input is valid
    #     students_score.append(num_student)

    ### Alternative 2
    # if num_student == 1 or num_student==2:
    #     students_score.append(num_student)
    # else:
    #     print("Invalid input:", num_student)

    ### Alternative 3
    if num_student not in [1,2]:
        print("Invalid input", num_student)
    else:
        students_score.append(num_student)

    """
    Ask for value again
    """
    num_student = int(input("Enter 1:for pass, 2: for fail: or (0: to exit): "))

# Variable to hold the totals
students_pass = 0
students_fail = 0

for students in students_score:
    if students == 1:
        students_pass += 1 
    else:
        students_fail +=1
        
# print("Total passed:", students_pass)
# print("Total failed:", students_fail)

if students_pass > 8:
    print("Congratulations students")
    print("Total passed:", students_pass)
    print("Total failed:", students_fail)
else:
    print("Total passed:", students_pass)
    print("Total failed:", students_fail)
```
