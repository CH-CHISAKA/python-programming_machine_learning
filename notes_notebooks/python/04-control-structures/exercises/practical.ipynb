# Project: Student Performance Management System

Problem statement

Develop a Python program that accepts student marks, calculates performance statistics, determines grades, and summarizes the class results.

**Requirements**

The program should:
1. Accept marks for multiple students.

2. Stop accepting marks when the user enters -1.

3. Validate marks to ensure they are between 0 and 100.

4. Determine each student's grade.

5. Calculate the class average.

6. Identify the highest and lowest marks.

7. Calculate the mean, median and mode.

8. Count the number of students who passed and failed.

9. Display a summary report.

```python
#### Import Libraries
import statistics

#### Initialze variables 

### Counters
students_passed = 0
students_failed = 0

### Store student grades
students_grades = []

"""
Get first input
"""

results = int(input("Enter student's grade or (-1 to exit): "))

while results != -1:
    '''
    Validate the enter values
    '''
    if results < 0 or results > 100:
        print("Invalid input")
        results = int(input("Enter student's grade or (-1 to exit): "))
    else:
        students_grades.append(results)
        '''
        Count the number of students who passed || failed
        '''
        if results > 60:
            students_passed += 1
        else:
            students_failed += 1

        """
        Ask for input again
        """
        results = int(input("Enter student's grade or (-1 to exit): "))


print("\n=========Class Summary Report=========")

"""
Identify the highest and lowest marks.
"""
print(f"Highest Grade: {max(students_grades)}")
print(f"Lowest Grade:  {min(students_grades)}")


"""
Calculate the Class Average
"""
print(f"Class Average: {sum(students_grades)/len(students_grades):.2f}")


"""
Calculate Mean, Median and Mode
"""
print(f"Class Mean:    {statistics.mean(students_grades):.2f}")
print(f"Class Median:  {statistics.median(students_grades):}")
print(f"Class Mode:    {statistics.mode(students_grades):}")


"""
Count the number of students who passed and failed.
"""

print("Students Passed:", students_passed)
print("Students Failed:", students_failed)
```

```python
import statistics

# Initialize variables
grades = []
passed = 0
failed = 0

print("STUDENT PERFORMANCE MANAGEMENT SYSTEM")
print("--------------------------------------")

# Collect student marks
while True:

    mark = int(input("Enter student marks (0-100), or -1 to finish: "))

    # Check sentinel value
    if mark == -1:
        break

    # Validate marks
    if mark < 0 or mark > 100:
        print("Invalid marks. Enter a value between 0 and 100.")
        continue

    # Store valid marks
    grades.append(mark)

    # Determine student performance
    if mark >= 60:
        passed += 1
        print("Result: Passed")

    else:
        failed += 1
        print("Result: Failed")


# Generate class report
print("\nCLASS PERFORMANCE REPORT")
print("------------------------")

if grades:

    total = sum(grades)
    count = len(grades)
    average = statistics.mean(grades)

    print(f"Number of students: {count}")
    print(f"Total marks: {total}")
    print(f"Class average: {average:.2f}")
    print(f"Highest mark: {max(grades)}")
    print(f"Lowest mark: {min(grades)}")
    print(f"Median: {statistics.median(grades)}")

    # Mode requires at least one value
    print(f"Mode: {statistics.mode(grades)}")

    print(f"Passed: {passed}")
    print(f"Failed: {failed}")

else:
    print("No valid student marks were entered.")
```

```python
import statistics

# Initialize variables
students_passed = 0
students_failed = 0
students_grades = []
letter_grades = []

# Get first input
results = int(input("Enter student's mark (or -1 to exit): "))

while results != -1:
    # Validate the entered values (0 to 100)
    if results < 0 or results > 100:
        print("Invalid input. Please enter a mark between 0 and 100.")
    else:
        students_grades.append(results)

        # Determine each student's letter grade
        if results >= 90:
            grade = "A"
        elif results >= 80:
            grade = "B"
        elif results >= 70:
            grade = "C"
        elif results >= 60:
            grade = "D"
        else:
            grade = "F"

        letter_grades.append(grade)
        print(f"Recorded: {results} -> Grade: {grade}")

        # Count the number of students who passed or failed
        if results >= 60:
            students_passed += 1
        else:
            students_failed += 1

    # Ask for input again
    results = int(input("Enter student's mark (or -1 to exit): "))

# Display summary report if at least one valid mark was entered
if len(students_grades) > 0:
    print("\n==============================")
    print("     CLASS SUMMARY REPORT     ")
    print("==============================")
    print(f"Total Students:  {len(students_grades)}")
    print(f"Highest Mark:    {max(students_grades)}")
    print(f"Lowest Mark:     {min(students_grades)}")
    print(f"Class Average:   {sum(students_grades)/len(students_grades):.2f}")
    print(f"Class Mean:      {statistics.mean(students_grades):.2f}")
    print(f"Class Median:    {statistics.median(students_grades)}")
    print(f"Class Mode:      {statistics.mode(students_grades)}")
    print(f"Students Passed: {students_passed}")
    print(f"Students Failed: {students_failed}")

    print("\n--- Individual Student Results ---")
    for i in range(len(students_grades)):
        print(f"Student {i + 1}: Mark = {students_grades[i]} | Grade = {letter_grades[i]}")
else:
    print("\nNo valid marks were entered.")
```
