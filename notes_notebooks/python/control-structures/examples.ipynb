# Example 1 (`if...else`)
You live in a country with only two tax brackets. Everyone earning less than 12,000 pays 25% in taxes, and anyone earning 12, 000 or more pays 30%.
* Calculate how much tax is owed.

```python
def get_tax(earnings):
    
    # check condition for earnings less than 12,000
    if earnings < 12000:
        tax_returns = earnings * 0.25
        
    # for earnings == or > than 12,000
    else:
        tax_returns = earnings * 0.3

    # return value for execution
    return tax_returns

# Call the function
print(get_tax(13000))
```

```python
# Compute for the following employees
ana_taxes = get_tax(9000)

bob_taxes = get_tax(15000)

print(f"Ana pays:$ {ana_taxes} in taxes")
print("Bob pays: $",bob_taxes,"in taxes")
```

In each case, we call the `get_tax()` function and use the value that is returned to set the value of a variable
* For ana_taxes - we calculate taxes owed by a person who earns `9,000`. In this case, we call the `get_tax()` function with earning set to `9,000`. Thus, `earnings is set to 9,000: earnings = 9,000 and evaluates earning < 12000 is True`, and tax returns is set to `9,000 * 0.25`. Then we return the value of `tax_returns`.

* For bob_taxes - we calculate taxes owed by a person who earns `15,000`. In this case, we call the `get_tax()` function with earnings set to `15,000`. Thus, `earnings is set to 15,000: earnings = 15,000 and ' earnings < 12000 ' evaluates to `False`, and tax returns is set to `15,000 * 0.3.` Then we return the value of `tax_returns`.

# Example 2 (`if...elif..else`)
Note that there is no limit to the number of `elif` one can use.

Calculate the does of medication (in milliliters) to give to a child, based on weight (in kilograms).

```python
def medication_dose(weight):
    
    # Dosage is 1.25 ml for anyone under 5.2 kg 
    if weight < 5.2:
        dose = 1.25
        
    elif weight < 7.9:
        dose = 2.5
        
    elif weight < 10.4:
        dose = 3.75
        
    elif weight < 15.9:
        dose = 5
        
    elif weight < 21.2:
        dose = 7.5

    # Dosage is 10 ml for anyone 21.2 kg or over
    else:
        dose = 10

    return dose

# Define weight
weight = 15

print("For weight", weight, "kg, provide a dosage of:",medication_dose(15), "ml")
    
```
