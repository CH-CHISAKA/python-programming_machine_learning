# Dictonaries

These are built-in Python data structures for mapping keys to values (_stores key-value pairs_)

* Store **key–value pairs**  
* Are **ordered** (since Python 3.7)  
* Are **mutable** (values can be added, deleted, or modified)  
* Keys must be **unique** and **immutable**  
* Key and value types **can differ**

**Examples:**

| Key | Key Type | Value | Value Type |
| :-- | :-- | :-- | :-- |
| Countries | `str` | Internet country codes | `str` |
| Decimal numbers | `int` | Roman numerals | `str` |
| States | `str` | Agricultural products | List of `str` |
| Hospital patients | `str` | Vital signs | Tuple of `int` and `float` |
| Metric dimensions | `str` | Abbreviations | `str` |
| Inventory codes | `str` | Quantity in stock | `int` |

## Creating a Dictionary

- A dictionary consists of `key-value` pairs in form of `key: value`, enclosed in curly braces `{}`
- One can create an empty dictionary by using `empty curly braces {}`

```python
{} # empty dictonary
```

```python
type({})
```

## Dictionary with key-value pairs

```python
country_codes = {'Kenya': 'KE', 'Switzerland': 'CHF', 'Sweden': 'SE', 'Finland': 'FI', 'South Africa': 'ZA', 'Nepal': 'NP'}
```

```python
country_codes
```

## Accessing the value associated with a key

```python
country_codes['Kenya']
```

```python
# Returns ['KE', 'ZA']
[country_codes[country] for country in ['Kenya', 'South Africa']]
```

## Checking if a Dictonary is Empty

```python
# Option 1: Length of the Dictionary
len(country_codes)
```

```python
# # Option 2: Dictionary as a Condition – True if the Dictionary is not empty
if country_codes:
    print("Dictionary not empty")
else:
    print("Dictionary is empty")
```

## Clear all the values in a Dictionary

```python
'''Using the `clear()` method to remove all the values in the dictionary'''
country_codes.clear()
```

```python
len(country_codes)
```

```python
'not empty' if country_codes else 'empty'
```

## Iterating 

```python
days_per_month = {'January': 31, 'February': 28, 'March':31}
```

```python
days_per_month
```

```python
for month in days_per_month:
    print(month)
```

```python
## The method `items()` returns each key-value pair as a tuple
days_per_month.items()
```

```python
for month, day in days_per_month.items():
    print(f"{month} has {day} days")
```

## Adding and Updating the Value of an Existing Key-Value Pair

```python
'''Adding a new value to an existing key-value pair'''
days_per_month['April'] = 30
```

```python
days_per_month
```

```python
for month in days_per_month.items():
    print(month)
```

```python
'''Updating the value of an existing key-value pair'''
days_per_month['February'] = 29
```

```python
days_per_month
```

## Removing the Value of a Key-Value Pair

```python
del days_per_month['January']
```

## Accessing Values using the get() method

- `get()` returns the value for the given key, or None if the key is not found.
- `get()` with a second argument returns that second argument if the key is not found.

```python
# Dictionary mapping Roman numerals to integers
roman_numerals = {'I': 1, 'II': 2, 'III': 3, 'V': 5, 'X': 5}
```

```python
roman_numerals
```

```python
roman_numerals.get('V')
```

```python
print(roman_numerals.get('V'))
```

```python
roman_numerals.pop('III')
```

```python
roman_numerals.get('III')  # None (not displayed in IPython)
```

```python
print(roman_numerals.get('III'))
```

```python
roman_numerals.get('III', 'III not in dict')
```

```python
roman_numerals.get('V')
```

## Dictionary Methods keys() and values()

```python
months = {'Jan': 1, 'Feb': 2, 'Mar': 3, 'Apr': 4}
months
```

#### The dictionary method `keys()` returns each key:

```python
months.keys()
```

```python
list(months.keys())
```

#### The dictionary method `values()` returns each key:

```python
months.values()
```

```python
list(months.values())
```

**Note:** The methods `items()`, `keys()`, and `values()` return a *view* of the data.  
A view provides access to the current contents of the dictionary — it does **not** store its own copy of the data.

## Comparing Dictionaries
Comparing two dictionaries returns `True` if both contain the same key–value pairs.  
The order of the key–value pairs does not matter.

```python
country_capitals1 = {'Belgium': 'Brussels', 'Haitiy': 'Port-au-Prince'}
```

```python
country_capitals2 = {'Nepal': 'Kathmandu', 'Uruguay': 'Monteviedeo'}
```

```python
country_capitals3 = {'Haitiy': 'Port-au-Prince', 'Belgium': 'Brussels'}
```

```python
country_capitals1 == country_capitals2
```

```python
country_capitals1 == country_capitals3
```

```python
country_capitals1 != country_capitals2
```

## Dictionary Method `update()`
The `update()` method can insert and update key–value pairs.

```python
country_codes = {}
```

```python
# Inserting a Dictionary
country_codes.update({'South Africa': 'za', 'Switzerland': 'ch'})
```

```python
country_codes
```

```python
# Intentionally inserting an incorrect key–value pair
country_codes.update({'Australia':'ar'})
```

```python
country_codes
```

```python
# Correcting the incorrect value
country_codes.update(Australia='au')
```

```python
country_codes
```

```python
# Inserting key–value pairs as a list of tuples
country_codes.update([('Italy', 'it'), ('Germany', 'ge')])
```

```python
# Inserting key–value pairs as a list of tuples
country_codes.update([('Italy', 'it'), ('Germany', 'ge')])
```

```python
country_codes
```
