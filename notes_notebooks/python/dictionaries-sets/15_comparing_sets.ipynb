# Comparing Sets

## Equality 

* Two sets are equal if they contain exactly the same elements.

* Since sets are unorded, the order of elements does not matter.

```python
{1, 3, 5} == {3,5,1}
```

```python
type({1, 3, 5} == {3,5,1})
```

```python
{1, 3, 5} != {3,5,1}
```

## Proper Subset

* The `<` operator tests whether the set on the left is a **proper subset** of the set on the right - that is, all elements of the left operand are contained in the right operand, and the **sets are not equal**.

```python
{1, 4, 8, 9, 11} < {9, 11, 4, 8, 1}
```

```python
{1, 9, 11} < {9, 11, 4, 8, 1}
```

## Improper Subset
* The `<=` operator tests whether the set on the left is an **improper subset** of the set on the right - that is, all elements of the left operand are also contained in the right operand, and the **sets may be equal**.

```python
{1, 4, 8, 9, 11} <= {9, 11, 4, 8, 1}
```

```python
{1, 9, 11} <= {9, 11, 4, 8, 1}
```

<br>

The set method **`issubset()`** also checks for an improper subset.  
The argument can be any iterable object, which will automatically be converted into a set.

```python
{1, 4, 8, 9, 11}.issubset({9, 11, 4, 8, 1})
```

```python
{1, 9, 11}.issubset([9, 11, 4, 8, 1])
```

## Proper Superset

```python
{1, 3, 5} > {3, 5, 1}
```

```python
{1, 3, 5, 7} > {3, 5, 1}
```

## Improper Superset

```python
{1, 3, 5} >= {3, 5, 1}
```

```python
{1, 3, 5} >= {3, 1}
```

```python
{1, 3} >= {3, 1, 7}
```

<br>

The set method **`issuperset()`** checks for an improper superset.  
The argument can be any iterable object, which will then be converted into a set.

```python
{1, 3, 5}.issuperset({3, 5, 1})
```

```python
{1, 3, 5}.issuperset([1, 5, 3])
```

```python
{1, 3, 5}.issuperset({3, 2})
```

```python
{'1', '3', '5'}.issuperset('135')
```
