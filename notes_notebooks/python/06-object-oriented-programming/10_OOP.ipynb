# PART 2: SPECIAL METHODS

## Understanding `__str__()` and `__repr__()`
Python provides special methods, commonly called dunder methods because their names start and end with double underscores.

Two important ones are:

- `__str__()`: Provides a readable representation for users.

- `__repr__()`: Provides a more informative representation for developers.

### The `__str__()` method
Called when you use print() on an object.

```python
class Account:

    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def __str__(self):
        return f"Account of {self.name}, balance: {self.balance}"
```

```python
### Create an instance of class Account

account1 = Account("John Doe", 1000)

### Output
print(account1)
```

Without `__str__()`, Python normally prints a default object representation.

### The `__repr__()` method
Provides a more formal representation.

```python
#### Create a class Account 

class Account:

    #### Initialize and arguments
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    ### Using __repre__() method
    def __repr__(self):
        return f"Account(name='{self.name}', balance={self.balance})"
```

```python
### Instantiate the class 
account2 = Account("Jane Doe", 5000)

### Output
account2
```

**Difference between them**

| Feature | `__str__()` | `__repr__()` |
| --- | --- | --- |
| Purpose | User-friendly output | Developer-friendly output |
| Called by | `print(object)` | `repr(object)` |
| Interactive notebook | Not normally the default | Normally displayed |
| Style | Readable description | Detailed representation |
