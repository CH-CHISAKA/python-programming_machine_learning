## Understanding Properties
Properties allow us to control how attributes are accessed and modified.

This is particularly important when working with sensitive information such as:
- Bank account balances.
- Employee salaries.
- Time values.
- Interest rates.
- Transaction fees.

Imagine a bank account with a balance of KES 10,000.

Without proper validation, someone could do this:


```bash 
account.balance -= 50000
```

This creates an invalid account balance.

- Properties help us prevent this.

## Public and internal attributes
Python uses naming conventions to indicate how attributes should be accessed.

| Attribute | Meaning |
| --- | --- |
| `name` | **Public** attribute – Anyone can access and modify it. |
| `_balance` | Intended for internal use (**protected** convention) – A polite warning to other developers: "Please don't access or modify this directly." |
| `__balance` | **Hidden** - Name-mangled attribute – Python changes the attribute name internally to reduce accidental access or modification. |

For example:

```bash 
self._balance = balance
```
We can then expose it through a property.


## Getter property - `@property`

A getter allows us to read an internal attribute through a normal-looking attribute name.

```bash
class Account:

    def __init__(self, balance):
        self._balance = balance

    @property
    def balance(self):
        return self._balance
```

````python
class Account:
    def __init__(self, balance):
        self._balance = balance


    
    ### Using the @property to access the protect variable
    @property
    def balance(self):
        return self._balance
````

````python
### Instantiation of the Account class
account = Account(1000)

print(account.balance)
````

Notice

```bash
account.balance
```

We do not write:
```bash 
account.balance()
```

Although `balance()` looks like a method, the `@property` decorator allows us to acces it like an attribute.

## Setter property – `@attribute.setter`

A setter allows us to control how an attribute is modified.

````python
class Account:

    def __init__(self, balance):
        self._balance = balance

    ### Getter for the protected attribute
    @property
    def balance(self):
        return self._balance

    ### Setter
    @balance.setter
    def balance(self, value):

        if value < 0:
            raise ValueError("Balance cannot be negative")

        self._balance = value
````

````python
### Instantiation
account2 = Account(5000)
print(account2.balance)

account2.balance = 8000
print(account2.balance)
````

**Creating a brand-new object VS Updating an existing one**

| Aspect | `account2 = Account(8000)` | `account2.balance = 8000` |
| --- | --- | --- |
| Action | Instantiation: Creates a completely new `Account` object in memory and makes `account2` point to it. | Attribute Mutation: Keeps the same `Account` object and simply overwrites its `balance` attribute. |
| What it does | Creates a new object. | Modifies an existing object's attribute. |
| `__init__()` method | Runs the class's `__init__()` constructor again from scratch. | Does not run `__init__()`. |
| Other attributes | Initializes other properties (like `account_number`, `owner`, or `transaction_history`) with their default or newly assigned values. | Preserves all other data already stored inside `account2`. |
| Memory & Identity | If the assignment creates a new object, `id(account2)` changes. | Keeps the same object in memory, so `id(account2)` stays the same. |


**Real-World Analogy**

* `account2 = Account(8000)` is like closing your bank account, throwing away the paperwork, and opening a brand-new account with `$8,000`.
* `account2.balance = 8000` is like keeping your exact same bank account and just changing the number on your balance sheet to `$8,000`.

### How does the setter work?

When you execute:

```bash 
account.balance = 8000
```

Python automatically calls:
```bash 
@balance.setter
def balance(self,value):
```
The supplied value becomes `value`.

The setter validates it before updating `_balance`.

## Read-Only vs Read-Write Properties

### Read-only property

```bash
@property
def balance(self):
    return self._balance
```
* You can read the balance.
* You cannot assign to account.balance.
* Changes must happen through methods such as deposit().

### Read-write property
```bash
@property
def balance(self):
    self._balance

@balance.setter
def balance(self, value):
    self._balance = value
```

* You can read the balance.
* You can assign a new balance.
* You can validate the new value before accepting it.

**Exam tip**: 

If an exercise requests a read-only property, implement @property without adding a setter.

## Complete Example: Bank Account with Properties
This combines:
- Classes.
- Objects.
- Constructors.
- Instance attributes.
- Getters.
- Validation.
- Methods.
- String representations.

````python

class Account:
    # Create a new bank account
    def __init__(self, name, balance=0):

        # Make sure the starting balance is not negative
        if balance < 0:
            raise ValueError("Initial balance must be >=0.")

        # Store the account holder's name and balance
        self._name = name
        self._balance = balance

    # Return the account holder's name
    @property
    def name(self):
        return self._name

    # Return the current account balance
    @property
    def balance(self):
        return self._balance

    # Add money to the account
    def deposit(self, amount):

        # Check that the deposit is greater than zero
        if amount <= 0:
            raise ValueError("Deposit must be positive.")

        # Increase the balance
        self._balance += amount

    # Remove money from the account
    def withdraw(self, amount):

        # Check that the withdrawal is greater than zero
        if amount <= 0:
            raise ValueError("Withdrawal must be positive.")

        # Check that there is enough money in the account
        if amount > self._balance:
            raise ValueError("Insufficient funds.")

        # Reduce the balance
        self._balance -= amount

    # Display the account information in a readable format
    def __str__(self):
        return (
            f"Account holder: {self.name} \n"
            f"Balance: {self.balance}"
        )

    # Show the account information for debugging
    def __repr__(self):
        return (
            f"Account(name= {self.name} )\n"
            f"Balance= {self.balance}"
        )
````

````python
# Create an account for John Doe with 50,000
account = Account("John Doe", 50000)

# Display the starting balance
print(account.balance)

# Deposit 8,000 into the account
account.deposit(8000)

# Display the updated account information
print(account)

# Withdraw 7,000 from the account
account.withdraw(7000)

# Display the final account information
print(account)
````

In your initial code, @property makes name and balance read-only attributes, meaning they can be viewed but not directly overwritten from outside the class.

__1. How name and balance work without a setter_

- In your `__init__` method, you created two internal variables with a leading underscore: `self._name` and `self._balance`.
   * In Python, a leading underscore signals that these variables are _private_ and **shouldn't be touched directly**.
  
- When you put `@property` above `def name(self)`: and `def balance(self`):, you create a `getter`:

- Reading works: When you run `print(account2.balance)`, Python runs the `balance()` method behind the scenes and returns `self._balance`.
    * It looks like a normal variable, even though it is actually calling a function.

- Writing is blocked: Because there is no `@balance.setter` or `@name.setter`, Python locks them. If someone tries to run `account2.balance = 500000`, Python throws an AttributeError: can't set attribute.

_2. Why we use methods to update the balance`

- If you added a standard `@balance.setter`, a user could simply overwrite the total balance with any number `(account2.balance = 1000000)`.
    * For a bank account, you don't want people arbitrarily replacing their balance—you want them to adjust it through controlled rules.
  
- Using methods like `deposit()` and `withdraw()` protects the data:
    * Enforces business rules: A method lets you check that deposits are positive and withdrawals don't exceed the current balance before `self._balance` changes.

- Performs math instead of replacement: A setter typically replaces an old value with a new one `(self._balance = value)`, whereas a method calculates a change based on the existing value `(self._balance += amount)`.

- Inside vs. Outside access: Even though balance is read-only from outside the class, methods inside the class are allowed to modify the private `self._balance` variable directly.
