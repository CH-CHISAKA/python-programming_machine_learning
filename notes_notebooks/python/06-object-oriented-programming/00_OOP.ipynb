# Introduction to Object-Oriented Programming 

Object-Oriented Programming (OOP) is a programming approach where we organise our code around **objects**.

## Objects in Software

* **Objects** model real world items, e.g `a car`, or `a bank account`
* We model objects in a class.
* A class is a blueprint or template for an object

An object represents something that has:

1. Data (what it knows).

2. Behaviour (what it can do).

## Objects Have a State and Behaviour

* **attributes** store the **values** *(data)* of an object

* **methods** define the **behaviour** of an object.
  They can:
  * Manipulate data
  * Perform claculations
  * Produce output (e.g, print the state)

The values of an object at a specific moment in time are called its **state**

| Concept | Role in OOP | Code vs. Runtime | Example (Car Class) |
| --- | --- | --- | --- |
| Attribute | The container or variable name (also called a field or property). | Defined in the class blueprint (static structure). | `color`, `speed`, `fuel_level` |
| State | The actual data or snapshot of values held by an instance. | Exists in memory at runtime and changes as the program runs (dynamic). | `color = "Red"`, `speed = 65`, `fuel_level = 40%` |

## Classes vs Objects


- A **class** is a blueprint or template used to create objects.

- An **object** is an actual instance created from that class.

| Concept | Description | Example |
| --- | --- | --- |
| Class | A blueprint that defines the attributes and methods that its objects will have. It specifies the rules but does not represent an actual dog. | `Dog` class with `name` and `breed` attributes and a `bark()` method. |
| Objects | Specific instances created from a class at runtime. Each object has its own attribute values. | `dog1`: `name = "Buddy"`, `breed = "Golden Retriever"` |
| Objects | Another instance of the same class with different attribute values. | `dog2`: `name = "Bella"`, `breed = "Beagle"` |

## Attributes and Methods

- **Attributes** store information about an object.

- **Methods** define what an object can do.

```python
# Create a class called Car
class Car:

    # Initialize the car with its brand and colour
    def __init__(self, brand, colour):
        self.brand = brand    # Store the car's brand
        self.colour = colour  # Store the car's colour

    # Define what happens when the car drives
    def drive(self):
        print(f"The {self.brand} is driving.")

    # Define what happens when the car stops
    def stop(self):
        print(f"The {self.brand} has stopped.")
```

Simple explanation

* `class Car`: – Creates a blueprint for making cars.

* `__init__` – Initializes a new car when it is created.

* `self.brand` – Stores the car's brand.

* `self.colour` – Stores the car's colour.

* `drive()` – Defines an action that makes the car drive.

* `stop()` – Defines an action that makes the car stop.

```python
car1 = Car("Toyota", "Red")  # Create an instance of Car
car2 = Car("Honda", "Blue")  # Create another instance of Car
```

```python
print(f"Car 1: {car1.brand}, Colour: {car1.colour}")  # Output the details of car1
print(f"Car 2: Colour: {car2.colour}")  # Output Colour details of car2
```

```python
car1.drive() # Call the drive method for car1
car2.stop() # Call the stop method for car2
```

## Abstraction and Modularization

* **Abstraction** is the ability to ignore details of certain parts to focus on a higher-level view of a problem.
* **Modularization** is the process of dividing a whole into well-defined parts that can be developed and examined independently and that interact with each other in a clearly defined manner.

* In Python, **attributes** are the **variables attached to an object**, while **state** is the **collective value of those attributes at any given moment**.

Think of a _video game character_: the **attributes** are things like _health_, _level_, and _weapon_. 

The character's **state** is the _exact data at a specific point in time_ —for instance, having _health = 80_, _level = 5_, and _weapon = "Sword"_. If they get hit, the health attribute drops to 50, meaning the character's state has changed.

In Python, **methods** are the **functions attached to a class**, while **behavior** is **what happens when you run those methods**.

Returning to the video game character analogy: 
* if _attributes_ are the _stats (like health)_ and _state is the snapshot of those stats (health = 80)_,
* then **methods** are the actions the character can perform **(like take_damage()** or **heal())**, and **behavior** is the **actual result—such as the character's health dropping** or a **"game over" screen triggering**.
