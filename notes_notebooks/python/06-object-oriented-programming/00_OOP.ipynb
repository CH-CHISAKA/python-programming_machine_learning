## Objects in Software

* **Objects** model real world items, e.g `a car`, or `a bank account`
* We model objects in a class.
* A class is a blueprint or template for an object

## Objects Have a State and Behaviour

* **attributes** store the **values** *(data)* of an object
* **methods** define the **behaviour** of an object.
  They can:
  * Manipulate data
  * Perform claculations
  * Produce output (e.g, print the state)

The values of an object at a specific moment in time are called its **state**

## Abstraction and Modularization

* **Abstraction** is the ability to ignore details of certain parts to focus on a higher-level view of a problem.
* **Modularization** is the process of dividing a whole into well-defined parts that can be developed and examined independently and that interact with each other in a clearly defined manner.

## Modeling with UML Class Diagrams

### Example: class `Song`
![Song](img/Song.png)

* In Python, **attributes** are the **variables attached to an object**, while **state** is the **collective value of those attributes at any given moment**.

Think of a _video game character_: the **attributes** are things like _health_, _level_, and _weapon_. 

The character's **state** is the _exact data at a specific point in time_ —for instance, having _health = 80_, _level = 5_, and _weapon = "Sword"_. If they get hit, the health attribute drops to 50, meaning the character's state has changed.

In Python, **methods** are the **functions attached to a class**, while **behavior** is **what happens when you run those methods**.

Returning to the video game character analogy: 
* if _attributes_ are the _stats (like health)_ and _state is the snapshot of those stats (health = 80)_,
* then **methods** are the actions the character can perform **(like take_damage()** or **heal())**, and **behavior** is the **actual result—such as the character's health dropping** or a **"game over" screen triggering**.
