## Class Definition

* A class definition begins with the keyword `**class**`, followed by the `class name` and a `colon (:)`.
  - This line is called the `class header`.

Python Style Guide (PEP 8) recommends writing class names in **PascalCase**, meaning:
  - A class name starts with a capital letter, and if it consists of multiple words, each word begins with a capital letter.
  - Example:
    * `CardDeck`
    * `CommissionEmployee`

* Every statement inside a class suite is indented.
* Each class should typically include a **docstring** that immmediately follow the _class header_.

```python
class Account:
    """Represents a bank account with an account 
    holder's name and balance."""
    pass
```

To view a class docstring in iPython, you can use the following statement:

```python
Account?
```

```python
type(Account) ## The data type of a class is `type`
```
