# Sets

A collection of unordered, duplicate-free values

* **Unordered**: no indices; no direct access to elements; no slicing
* **Duplicate-free**
* **Mutable**: elements can be added or removed, but not modified.
* **Elements**: must be **immutable**
* **Iterable**

**Applications**:

* Removing duplicates.
* Set theory: comparing sets and creating new sets from existing ones.

## Creating an Empty Set 
* To create an empty set, one must use the **`set()` function**, since **`{}` represents an empty dictionary**.

* Python displays an empty set as `set()` to avoid confusion with an empty dictionary `({})`

```python
numbers = set()

type(numbers)
```

```python
numbers
```

## Set Comprehensions

```python
alt_numbers = set(list(range(10)))
```

```python
alt_numbers
```

```python
type(alt_numbers)
```

```python
even = set()

for item in alt_numbers:
    if item % 2 == 0:
        even.add(item) # .add() → adds an element to a set.
```

```python
even
```

```python
evens = {items for items in alt_numbers if items % 2 == 0}
```

```python
evens
```

## Frozenset: An Immutable Set Type
* **Sets are mutable.**  
* **Set elements must be immutable** — therefore, a set cannot contain other sets as elements.  
* A **frozenset** is an *immutable* set — it cannot be modified after creation.  
  Because of this, a set can contain *frozensets* as elements.  
* The built-in **`frozenset()`** function creates a frozenset from any iterable object.


```python
mutable_set = set('123')
```

```python
mutable_set
```

```python
mutable_set.add(4)
```

```python
mutable_set
```

```python
mutable_set.update([6])
```

```python
mutable_set
```

```python
my_fset = frozenset(mutable_set)
```

```python
my_fset
```

```python
my_fset.add(5)
```

```python
my_fset.update([4])
```

```python
my_fset.clear()
```
