# 3.2 Functions with Multiple Parameters

The following `maximum` function returns the largest of three values.

```python
def maximum(value1, value2, value3):
    """Return the maximum of three values."""
    max_value = value1

    if value2 > max_value:
        max_value = value2

    if value3 > max_value:
        max_value = value3

    return max_value
```

```python
maximum(12, 27, 36)
# maximum?
```

```python
maximum(12.3, 45.6, 9.7)
```

```python
maximum('yellow', 'red', 'orange')
# maximum?
```

```python
maximum('Yellow', 'red', 'orange')
# maximum?
```

## Python’s Built-in min and max Functions
* Frequently needed functions are already available in Python, such as `min` and `max`.
* Advantages: shorter development time, high reliability, and excellent performance.

```python
max('yellow', 'red', 'orange', 'blue', 'green')
```

```python
min(15, 9, 27, 14)
```
