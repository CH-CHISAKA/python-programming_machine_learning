# 3.4 Keyword Arguments

```python
def rectangle_area(length, width):
    """Return a rectangle's area."""
    return length * width
```

When calling a function, you can use **keyword arguments** to pass arguments in **any order**.

```python
rectangle_area(width=5, length=10)
```

```python
def cuboid_volume(length, width, height):
    """Calculate a cuboid volume."""
    return length * width * height
```

Keyword arguments must always appear **to the right of positional arguments** — those for which no parameter names are specified.

```python
cuboid_volume(length=5, 3, 2)
```

```python
cuboid_volume(5, height=2, width=3)
```

Keyword arguments can improve the readability of function calls that have many arguments.
