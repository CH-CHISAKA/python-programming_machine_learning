# 3.9 Methods  
* A **method** is a function that belongs to an object.  
* A method can be called using the following syntax:  
  <i><b>object_name.method_name(*args)</b></i>

**Example:**  
* The string `"Hello"` is a `str` object assigned to the variable `s`. 
* Then, the `str` methods `lower()` and `upper()` are called using dot (`.`) notation, each producing a **new string**.

```python
s = 'Hello'
s.lower()  # call lower method on string oject s
```

```python
s.upper()
```

```python
s
```

See the [Python Standard Library](https://docs.python.org/3/) for a list of available types and their methods.

## The Core Difference Between Methods and Functions  

* A **method** is a function that is *bound* to an object and can access or modify that object’s data.  
* A **function** is not bound to any specific object and operates only on the data explicitly passed to it.  

| Concept | Belongs to | How it’s called | Example |
|:---------|:------------|:----------------|:----------|
| **Method** | An **object** (instance of a class) | `object_name.method_name()` | `"Hello".upper()` |
| **Function** | A **module** or **the global namespace** | `function_name()` or `module_name.function_name()` | `len("Hello")`, `math.ceil(1.1)` |
