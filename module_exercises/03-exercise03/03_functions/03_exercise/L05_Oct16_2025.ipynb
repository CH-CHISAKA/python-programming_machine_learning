## Non-Mutable Objects as Arguements


````python
x = 7
print(f"id(x): {id(x)}")

def square(number):
    print(f"id(number): {id(number)}")
    number **= 2
    print(f"id(number) after calculating the square: {id(number)}")
    return number



square(x)   # after execution of this statement we will have 49 returned
print(f"id(x) after function call: {id(x)}")
print(x)
````

# Craps
* we roll two dice
* You win if on first roll if the sum of the two dice is 7 or 11
* You loose if on first roll if the sum of the two dice is 2, 3, or 12
* else: you memorize your **points** (=sum of the two dice)
* then you repeadetly roll the dice:
    * if you roll your point again: you win
    * if you roll a 7 before your point: you loose 

#### Pseudocode
```text
1. roll two dice
2. display the two dice and the sum of them 
3. determine game status (WON (7, 11), LOST (2, 3, 12), CONTINUE)

if CONTINUE:
    
    record point_value = result

    continue rolling until player wins or looses:

        roll two dice
        dispay result
        compare --> 7 = LOST, point_value --> WON

Display outcome
```

````python
import random
from enum import Enum
from typing import Tuple


class Status(Enum):
    WON = "WON"
    LOST = "LOST"
    CONTINUE = "CONTINUE"


def roll_dice() -> Tuple[int, int]:
    """Roll two dice and return their face values as a tuple (die1, die2)"""
    die1 = random.randint(1, 6)
    die2 = random.randint(1, 6)
    return (die1, die2)


def display_dice(dice: Tuple[int, int]) -> None:
    """Display one roll of two dice."""
    die1, die2 = dice
    total = die1 + die2
    print(f"Player rolled {die1} + {die2} = {total}")


def determine_first_roll_game_status(total: int) -> Status:
    """Return the game status aftet the come-out roll."""

    if total in (7, 11):
        return Status.WON
    if total in (2, 3, 12):
        return Status.LOST
    return Status.CONTINUE


def main():
    print("Welcome to Craps\n")

    dice = roll_dice()
    display_dice(dice)
    total = sum(dice)

    status = determine_first_roll_game_status(total)

    if status == Status.CONTINUE:
        point = total
        print(f"Point is {point}")

        while True:
            dice = roll_dice()
            display_dice(dice)
            total = sum(dice)

            if total == point:
                status = Status.WON
                break
            
            if total == 7:
                status = Status.LOST
                break

    ## display outcome
    print("Player wins" if status == Status.WON else "Player loses")
                

if __name__ == "__main__":
    main()
````
