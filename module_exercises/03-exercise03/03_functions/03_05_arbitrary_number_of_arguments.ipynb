# 3.5 Functions with an Arbitrary Number of Arguments

You can define functions that can receive an arbitrary number of arguments.

```python
def average(*args):
    """Returns the average value of the provided arguments."""
    return sum(args) / len(args)
```

```python
average(5, 10)
```

```python
average(5, 10, 15)
```

```python
average(5)
```

```python
average()
```

&nbsp;

## Enforcing That at Least One (or More) Arguments Must Be Passed

By modifying the previous average function as shown below, we now enforce that at least **two or more arguments** must be provided when calling the function.

```python
def average(arg1, arg2, *args):
    """Returns the average value of the provided arguments"""
    return (arg1 + arg2 + sum(args)) / (len(args) + 2)
```

```python
average()
```

```python
average(10, 5)
```

```python
average(10, 5, 15)   # 15 is packed into a tuple and passed to args
```

```python
average(10, 5, 15, 30)   # 15 and 30 are packed into a tuple and passed to args
```

&nbsp;
## Passing the Individual Elements of an Iterable Object

```python
grades = [10, 10, 20, 40]
average(*grades)
```

When the `*` operator is applied to an iterable argument in a function call, its elements are **unpacked**.

In other words, the call above is equivalent to:

```python
average(10, 10, 20, 40)
```

&nbsp;

## Final Remarks
* The parameter name *args* is used by convention, but you can choose any valid identifier.
* If a function has multiple parameters, the *\*args* parameter must appear **to the right** of the other parameters.
