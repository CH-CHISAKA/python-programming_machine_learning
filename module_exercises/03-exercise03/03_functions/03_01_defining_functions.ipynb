# 3.1 Defining Functions

* A function consists of:

    * the `def` keyword, followed by the function name and parentheses;
    * optional parameters inside the parentheses;
    * a *suite* - an indented group of statements forming the function body;
    * optionally, a `return` statement that usually provides a result.
<br>

* Function names always start with a lowercase letter and use *snake_case*.

* According to **PEP 8** (the *Style Guide for Python Code*), the first line inside a function’s suite should be a **docstring** that briefly describes the purpose of the function.

There are three ways a function can terminate:

1. A return statement **with a result**.
3. A return statement **without a result** → returns `None`.
4. **No** return statement at all → implicitly returns `None`.

When a function finishes executing, it **returns control to the caller**.

## 2. Function That Contains a return Statement Without a Result

It explicitly returns `None`.

```python
def return_value_demo():
    """Retune None"""
    return
```

```python
print(return_value_demo())
```

```python
return_value_demo()  # Returns None, but the result is not displayed here
```

### 3. Function Without a return Statement

It implicitly returns `None`.

```python
def no_return_demo():
    """No return statement returns None implicitly."""
    pass
```

### Function names are variables

```python
def function_name_demo():
    print('Function names are variables that refer to a function definition')
```

The following snippet (just the function name) will return the **function object** itself, not its result, i.e., it will not execute the function.

* \_\_main\_\_ indicates that the function is defined in the currently running module (your notebook or script)
* function_name_demo() is the **qualified name** of that function object.

```python
function_name_demo
```

```python
type(function_name_demo)
```

Adding parentheses to the function name calls (executes) the function.

```python
function_name_demo()   # The parentheses (with or without arguments) tell Python to execute the function
```
