# 3.8 Case Study: “Craps”

“Craps” is a popular dice game. In this section, we’ll **simulate the game in Python**.

## Excursus: Random Numbers  

The built-in `random` module provides functionality for working with random numbers.  
* To use it, you must first **import** the module.  
* The function `randint(a, b)` returns a random integer `n` such that `a <= n <= b` —  
  that is, both the lower and upper bounds are **included**.

&nbsp;
### Simulating a Die:

```python
import random

for roll in range(10):
    rolled_value = random.randint(1, 6)  # A die can randomly roll values from 1 to 6
    print(rolled_value, end='   ')
```

&nbsp;
### Reproducible “Random” Numbers Using the `seed` Function  

The `randint` function actually generates **pseudorandom numbers**, based on an internal calculation that starts from a numeric value called the **seed**.  

Repeated calls to `randint` produce a sequence of numbers that *appear* random.  
Each time you generate random numbers, Python uses a **different seed** by default, which results in a different sequence.  

When debugging programs that rely on randomly generated data, it can be useful to reproduce the **same sequence** of random numbers until all logic errors are fixed.  
You can achieve this by using the **`seed` function** from the `random` module.  

Setting a seed forces `randint` to start from a fixed initial value, ensuring that the pseudorandom sequence is **always identical** for that seed.

```python
import random

random.seed(1)  # Initialize the pseudorandom number generator with a fixed seed

# Generate the first sequence of random die rolls
for roll in range(10):
    print(random.randint(1, 6), end=' ')

print()

# This loop continues the same pseudorandom sequence
for roll in range(10):
    print(random.randint(1, 6), end=' ')
```

## The Game of “Craps”  

### Rules  

Two dice are rolled simultaneously. When they come to rest, the **sum** of the numbers on the upward-facing sides is calculated.  

* **You win** if the sum on the first roll is **7 or 11**.  
* **You lose** if the sum on the first roll is **2, 3, or 12** (known as *“craps”*).  
* **In all other cases**, the **point value** is recorded, and you continue rolling until the outcome is determined:  
  * **You win** if you roll the **same point value again**.  
  * **You lose** if you roll a **7** before rolling your point value.

#### Version 1

```python
import random

def roll_dice():
    """Roll two dice and return their face values as a tuple (die1, die2)."""
    die1 = random.randint(1, 6)
    die2 = random.randint(1, 6)
    return die1, die2  # pack into a tuple

def display_dice(dice):
    """Display one roll of the two dice."""
    die1, die2 = dice   # unpack the tuple
    total = die1 + die2
    print(f'Player rolled {die1} + {die2} = {total}')


# First roll
die_values = roll_dice()
display_dice(die_values)

# Determine game status and point, based on first roll
sum_of_dice = sum(die_values)

if sum_of_dice in (7, 11):          # win
    game_status = 'WON'
elif sum_of_dice in (2, 3, 12):     # lose (craps)
    game_status = 'LOST'
else:                                # establish point
    game_status = 'CONTINUE'
    my_point = sum_of_dice
    print('Point is', my_point)

# Continue rolling until player wins or loses
while game_status == 'CONTINUE':
    die_values = roll_dice()
    display_dice(die_values)
    sum_of_dice = sum(die_values)

    if sum_of_dice == my_point:      # player made the point
        game_status = 'WON'
    elif sum_of_dice == 7:
        game_status = 'LOST'

# Display outcome
if game_status == 'WON':
    print('Player wins')
else:
    print('Player loses')
```
