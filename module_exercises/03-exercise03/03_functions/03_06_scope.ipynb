# 3.6 Scope: Rules of Visibility
* Scope refers to the visibility or lifetime of variables, functions, and other identifiers.
* Every identifier (i.e., variable name, function name, etc.) has a well-defined visibility within the program.

## Local Variables
* Parameters and variables defined within a function’s suite are **local variables**.
* Attempting to access a local variable outside the function’s suite raises a `NameError`.
* The scope of these parameters and variables is **local**.


````python
def local_var_demo(number):
    """Demo function to demonstrate local variables."""
    a = 5
    return number + a
````

````python
local_var_demo(5)
````

````python
a
````

````python
number
````

## Global Variables  
* Variables defined outside a function (or class) have **global visibility**.  
* Global variables can be used anywhere in a `.py` file or in an interactive session **after** they have been defined.

````python
global_number = 10

def global_var_demo():
    """Demo function to demonstrate global variables."""
    return global_number + 10


global_var_demo()
````

````python
global_number
````

&nbsp;

## Modifying Global Variables Inside Functions

By default, global variables are **not modified** inside a function.

````python
global_number = 10

def modify_global_var_demo():
    """Demo function to demonstrate global variables."""
    global_number = 5

    return global_number


modify_global_var_demo()
````

````python
global_number
````

### The `global` Statement

````python
x = 10

def modify_global():
    global x
    x = 5
    print('x printed from modify_global', x)

x
````

````python
modify_global()
````

````python
x
````

## Scope
* A **suite** inherits its scope from the suite it belongs to:  
  * If the suite is part of the **global suite** (e.g., top of a script or notebook), its variables have **global scope**.  
  * If the suite is part of a **function suite**, its variables have **local scope**.

````python
x = 10

def local_scope_demo():
    """A suite inside a function block has local scope."""
    x = 1

    if x < 4:
        x = 4  # x has local scope here

    print(f'x in local scope: {x}')


local_scope_demo()

print(f'x in global scope after function call: {x}')


if x > 5:
    x = 5  # x has global scope here

print(f'x in global scope after if-statement: {x}')
````

## Function Shadowing  
⚠️ **Warning:** If you define a variable with the same name as a function, you can no longer call that function.

For example:

````python
sum([10, 5])   # built-in function sum
````

````python
sum = 100 + 200
sum
````

````python
sum([1, 2])
````

You can use the `del` statement to remove the variable `sum` from the current scope. This makes the built-in function `sum` callable again.

````python
sum
````

````python
type(sum)
````

````python
del sum
````

````python
sum
````

````python
sum([1, 2])
````

### Common Built-in Names Often Shadowed by Variables  

* `min`, `max`, `sum`, `list`  
* `dict`, `set`, `tuple`, `str`, `int`, `float`, `bool`  
* `len`, `range`, `input`, `open`, `file` (in older versions)  
* `id`, `type`, `dir`, `help`  
* `format`, `sorted`, `reversed`, `map`, `filter`, `zip`  
* `any`, `all`, `abs`, `round`  
* `object`, `super`  

> **Tip:** To check whether a name is already used by Python, you can inspect  
> `dir(__builtins__)` or import the `builtins` module:
> ```python
> import builtins
> dir(builtins)
> ```
