# 3.3 Default Parameter Values

A parameter can have a default value.

```python
def rectangle_area(length=2, width=3):
    """Return a rectangle's area."""
    return length * width
```

When calling the function, parameters with default values can be omitted — as long as no new arguments are meant to replace them.

```python
rectangle_area()  # No arguments are passed → default values are used
```

```python
rectangle_area(10)  # Arguments are assigned to parameters from left to right
```

```python
rectangle_area(10, 5)  # All parameters receive new argument values
```

### Parameters with default values must appear to the right of parameters without default values in the parameter list.

```python
def cube(length=1, width=1, height):
    """Calculate the cube."""
    return length * width * height
```

```python
def cube(height, length=1, width=1):
    """Calculate the cube."""
    return length * width * height
```

```python
cube(10)
```

```python
cube(10, 10, 10)
```
