# 3.7 Passing Arguments to Functions  
* In Python, arguments are always passed **by reference** (“pass by reference”).  
* When a function is called, Python copies the **reference** of the argument object into the corresponding parameter.

## Memory Addresses and References
&nbsp;
<img src="img/reference.png" width='250'>

A variable does not store the object itself, but a **reference** (a memory address) that points to the object in memory.

## The `id` Function  
Returns a unique integer that identifies the object’s location in memory.


```python
x = 7
id(x)   # Identity of the object — an integer value that references the object's location in memory
```

&nbsp;

## Passing an Object to a Function

```python
def cube(number):
    print('id(number):', id(number))
    return number ** 3
```

```python
cube(x)
```

The identity shown for the parameter `number` is the same as for `x`.  
This means that the argument `x` and the parameter `number` refer to the **same object** while `cube` is executing.  
In other words, `x` and `number` reference the same object in memory.

&nbsp;

## Testing Object Identities with the `is` Operator  
The `is` operator returns `True` if both operands have the same identity — that is, if they reference the **same object** in memory.

```python
x = 7
id(x)
```

```python
def cube(number):
    print('number is x:', number is x)  # x is a global variable
    return number ** 3

cube(x)
```

&nbsp;

## Non-Mutable Objects as Arguments  
* Objects of types such as `str`, `int`, and `float` are **immutable**.  
* When a reference to an immutable object is passed to a function, the original object is **not modified**.

```python
x = 7
id(x)
```

```python
def cube(number):
    print(f'id(number) before modifying number: {id(number)}')
    number **= 3
    print(f'id(number) after modifying number: {id(number)}')
    return number
```

```python
cube(x)
```

```python
id(x)
```

&nbsp;

## Mutable Objects as Arguments  
* Lists are **mutable** objects.  
* When a reference to a mutable object is passed to a function, the original object **can be modified** within the function.

```python
my_list = [1, 2, 3]
print(f'my_list before function call: {my_list}')

def list_mutation_demo(param_list):
    param_list.append(4)
    print(f'my_list is param_list? {my_list is param_list}')
    print(f'param_list: {param_list}')

list_mutation_demo(my_list)
print(f'my_list after function call: {my_list}')
```
