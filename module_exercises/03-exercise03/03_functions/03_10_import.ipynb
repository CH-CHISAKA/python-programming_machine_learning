# 3.10 Import: A Closer Look  
* Python groups related functions, data, and classes into **modules**.  
* Accessing these elements usually requires an **import** statement.

## Importing a Module  

You can import a module using the following statement:  

`import module_name`  

You can then access its functions using the module name, followed by a dot (`.`), and the function name you want to use.  

For example:

```python
import math

x = math.ceil(1.1)  # return the smallest integer greater than or equal to x

print(x)
```

## Importing a Specific Identifier from a Module  

If you want to import a specific identifier from a module, you can do so with a statement like this:  

`from module_name import identifier`  

You can then use the identifier directly in your code **without** prefixing it with the module name.  

For example:

```python
from math import ceil

x = ceil(1.1)

print(x)
```

## Importing Multiple Identifiers from a Module  

Using the `from ... import` statement, you can import multiple identifiers from a module, separated by commas.  
Afterward, you can use them directly in your code **without** prefixing the module name and a dot (`.`):

```python
from math import ceil, floor

ceil(10.3)
```

```python
floor(10.7)
```

Attempting to use a function that was **not imported** will cause a `NameError`, indicating that the name is not defined.

## Caution: Avoid Wildcard Imports  

You can import *all* identifiers defined in a module using a wildcard import of the following form:  

`from module_name import *`  

However, this can lead to subtle bugs if your own variable names are the same as function names in the module (**name shadowing**).  

Example:

```python
e = 'Emil'

from math import *

e  # The wildcard import has overwritten your variable 'e'
```

## Aliases for Modules or Identifiers  

Sometimes it is helpful to use an **alias** for modules or identifiers.

```python
import statistics as stats

grades = [85, 93, 45, 87, 93]

stats.mean(grades)
```

```python
from statistics import mean as m

grades = [10, 20]
m(grades)
```
