## Task 1
The following code is intended to display the unique words from `text` and the number of occurrences of each word.  

What is wrong with this code?

```python
from collections import Counter
text = ('to be or not to be that is the question')
counter = Counter(text.split())
for word, count in sorted(counter):
    print(f'{word:<12}{count}')
```


````python
from collections import Counter
text = ('to be or not to be that is the question')
counter = Counter(text.split())
for word, count in sorted(counter.items()):
    print(f'{word:<12}{count}')
````

## Task 2
The following dictionary represents a lecturer’s grade book, where each student’s scores from various exams are stored.

```python
grade_book = {            
    'Susan': [92, 85, 100], 
    'Eduardo': [83, 95, 79],
    'Azizi': [91, 89, 82],  
    'Pantipa': [97, 91, 92] 
}
```

Write a script that calculates and displays the average score per student as well as the class average.

````python
# Dictionary containing key: value pair

grade_book = {            
    'Susan': [92, 85, 100], 
    'Eduardo': [83, 95, 79],
    'Azizi': [91, 89, 82],  
    'Pantipa': [97, 91, 92] 
}

# importing libraries 
from statistics import mean

# Calculate average per student
avg_grades = {}

for key, values in grade_book.items():
    avg_grades[key] = mean(values)

# print(avg_grades) ## Test

# Display each student's average
for key, value in avg_grades.items():
    print(f'{key}: {value: .2f}')

#  Calculate the class average
class_avg = mean(avg_grades.values()) 

# class_avg ## Test output

print(f'Class Average {class_avg:.2f}')
````
