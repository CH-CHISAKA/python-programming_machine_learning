## Task 8
Display the diameter, circumference, and area of a circle with a radius of 2.

Use the following formulas:

* diameter = 2 · r
* circumference = 2 · π · r
* area = π · r²

where `r` is the radius.

**Hint:** Python's `math` module provides the constant π. Make the module available at the beginning of your program with:

```python
import math
```

You can then access π as `math.pi`.


````python
import math

# Calculating the diameter of a circle given its radius

# Radius of the circle
radius = 2

# Calculate the diameter using the formula: diameter = 2 * radius
diameter = 2 * radius

# Print the diameter
print("The diameter of the circle with, radius", radius, "is:", diameter)

# Calculate the circumference using the formula: circumference = 2 * π * radius
circumference = 2* math.pi *radius 

# Print the circumference
print(f"The circumference of the circle with radius, {radius}, is:, {circumference}")

# Calculate the area using the formula: area = π * radius^2
area = math.pi * radius ** 2

# Print the area
print(f"The area of the circle with radius, {radius}, is:, {area}")
````

## Task 10
Write a program that reads three integers from the user and then outputs the sum, average, product, smallest, and largest of the numbers.

````python
# User enters the first digit
input_1 = int(input("Enter the first digit:"))

# User enters the second digit 
input_2 = int(input("Enter the second digit:"))

# User enters the third digit
input_3 = int(input("Enter the third digit:"))

# Computing the sum of all the inputs from the user
sum_inputs = input_1 + input_2 + input_3 

# Print out the sum of the user inputs
print("The sum of the inputs is:", sum_inputs) 

# Computing the average of all the inputs from the user 
# avg_inputs = (input_1 + input_2 + input_3)/3
avg_inputs = (sum_inputs)/3

# Print the average input of the user 
print("The average of the input is:", avg_inputs)

# Calculate the product of the inputs
prd_inputs = input_1 * input_2 * input_3

# Print the product of the inputs
print(f"The product of all the inputs is: {prd_inputs}")

# Calculate the smallest interger input 
min_input = min(input_1, input_2, input_3)

# Print the smallest interger input
print(f"The smallest number entered is: {min_input}")

# Calculate the largest integer input
max_input = max(input_1, input_2, input_3)

# Print the largest integer input
print("The largest number entered is:", max_input)
````

<!-- Write a program that reads three floating point numbers and then outputs them in ascending order on the screen. -->

````python
# Asking the user to enter a range 
range_amount = int(input("Enter the amount of times you would like to run this program: "))

# Creating an empty list to store all the inputs
numbers = []

for i in range(range_amount):
    numbers.append(float(input("Enter a number: ")))
    
# Sort the entered numbers in ascending order using the `sort()` function which sorts in ascending order by default.
numbers.sort()

# Print the output
print("Printing the numbers in ascending order:")
for n in range(range_amount):
    print(numbers[n])

# print(numbers[0])
# print(numbers[1])
# print(numbers[2])
````
