## Experiment 5 - Is it a leap year? (boolean operators)

A year is a leap year if it is divisible by 4 - **except** if it is divisible by 100,
**unless** it is divisible by 400. (So 2000 is a leap year, 1900 is not, 2024 is.)

**Task:** Read a year and decide whether it is a leap year.

**Let's talk:**
- Write the rule as a *single boolean expression* with `and` / `or` / `not`.
- Which operator has higher precedence: `and` or `or`?

**Challenge:** print all leap years between 2000 and 2100.

````python
year = int(input('Enter a year:'))

is_leap = (year % 4 == 0) and (year % 100 != 0 or year % 400 == 0)

if is_leap:
    print(year, 'is a leap year')

else:
    print(year, 'is not a leap year')

for year in range(2000, 2100):
    is_leap = (year % 4 == 0) and (year % 100 != 0 or year % 400 == 0)
    if is_leap:
        print(year)
````

## Experiment 6 - Sentinel-controlled statistics

Read numbers from the user. Stop when the user enters `0`. Then display the **sum**,
the **count**, the **average**, and the **largest** value entered.
(Do not count the sentinel value `0` in the statistics.)

**Let's talk:**
- Which loop fits a problem where we don't know the number of inputs in advance?
- Where do we update `total` and `count`?
- How do we keep track of the largest value seen so far?

````python
# Initialize our tracking variables
total_sum = 0
count = 0
largest = None

print("Enter numbers one by one. Enter 0 to stop.")

while True:
    # 1. Read input from the user
    user_input = int(input("Enter a number: "))
    
    # 2. Check for the sentinel value immediately
    if user_input == 0:
        break  # Exit the loop entirely
        
    # 3. Update total and count
    total_sum += user_input
    count += 1
    
    # 4. Keep track of the largest value
    if largest is None or user_input > largest:
        largest = user_input

# 5. Calculate statistics and display results (only if at least one valid number was entered)
if count > 0:
    average = total_sum / count
    print("\n--- Statistics ---")
    print("Sum:", total_sum)
    print("Count:", count)
    print("Average:", average)
    print("Largest Value:", largest)
else:
    print("No numbers were entered before stopping.")
````

## Experiment 7 - mean, median and mode with the `statistics` module

The `statistics` module offers the functions `mean`, `median`, and `mode`.

**Task:** Given the following list of exam scores:

```python
scores = [62, 58, 74, 62, 91, 74, 62, 85]
```

display the mean, the median, and the mode.

**Let's talk:**
- Before running: *predict* the three results - how does the median differ from the mean here?
- What happens if we call `mode` on a list where *every* value occurs equally often?

**Experiment idea:** find a second list where `mode` raises an error, and explain why.

````python
import statistics

# Given list of exam scores
scores = [62, 58, 74, 62, 91, 74, 62, 85]

# Calculate statistics
mean_value = statistics.mean(scores)
median_value = statistics.median(scores)
mode_value = statistics.mode(scores)

# Display the results
print(f"Mean:   {mean_value}")
print(f"Median: {median_value}")
print(f"Mode:   {mode_value}")
````

* The Mean (71) is higher because it calculates the mathematical average. High scores like 85 and 91 pull the mean upward.

* The Median (68.0) is lower because it represents the exact midpoint of the data. When the list is sorted ([58, 62, 62, 62, 74, 74, 85, 91]), the two middle numbers are 62 and 74. The median is the average of those two numbers: (62 + 74) / 2 = 68.0.


## Experiment 8 - break and continue

Two small experiments with the loop "jump" statements.

1. Print the numbers 1 to 20, but **skip** all multiples of 5 (`continue`).
2. Print the numbers 1 to 20, but **stop** as soon as you reach a multiple of 6 (`break`).

**Let's talk:**
- What is the difference between skipping an iteration and ending the loop?
- In the second part: which numbers actually get printed?

**Challenge:** swap the roles - *break* on multiples of 5 and *continue* on multiples of 6.

````python
print("--- Experiment 1: Skipping Multiples of 5 ---")
for i in range(1, 21):
    if i % 5 == 0:
        continue  # Skip the rest of this iteration
    print(i, end=" ")
print("\n")

print("--- Experiment 2: Stopping at Multiples of 6 ---")
for i in range(1, 21):
    if i % 6 == 0:
        break     # Exit the loop completely
    print(i, end=" ")
print("\n")
````
