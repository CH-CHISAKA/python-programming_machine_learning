## Task 1

Modify the following script to validate its input.
If the user enters any value other than 1 or 2, the input should be ignored and the user should be prompted to enter a new value.
Use a counter to keep track of the number of invalid entries.
When the program ends, display the number of invalid entries along with the existing output.

```
PASS = 1
FAIL = 2

passed = 0
failed = 0

for student in range(1, 11):  
    result = int(input(f'Enter the result for student {student} (1 = passed, 2 = failed): '))

    if result == PASS:
        passed += 1
    else:
        failed += 1
        
print(f'Passed: {passed}')
print(f'Failed: {failed}')
```

````python
# Accepted values from user
PASS = 1
FAIL = 2

# Counter variable to track how many times the user's invalid entries 
invalid = 0

# Counter to track passed and failed students
passed = 0
failed = 0

# Results for 10 students
for student in range(1,11):
    # For invalid user input, keep looping 
    while True:
        result = int(input(
            f"Enter the result for student: {student},"
            f"(1 = passed, 2 = failed): "
            )
        )
    
        # Validation and error handling function (check the value)
        if result == PASS:
            passed+=1
            break
        
        elif result == FAIL:
            failed+=1
            break
        
        else:
            # invalid number in counter 
            invalid+=1
            print("Invalid input, Please enter 1 or 2.")
    
# Results
print(f'Passed: {passed}')
print(f'Failed: {failed}')
print(f'Invalid: {invalid}')
````

For example, the number 54823 should be displayed as:

```
5
4
8
2
3
```
Derive the individual digits using the operators // and %.

````python
# Invalid user entries count
invalid = 0

# 
while True:
    user_input = input("Enter a five digit number: ")

    # Check the input is 5 digits long 
    if len(user_input) == 5 and user_input.isdigit():
        user_input = int(user_input)
        break

    # Handle invalid inputs
    else:
        invalid+=1
        print("Invalid inputs:")

# Derive the individual digits using the operators // and %.
digit1 = user_input // 10000
digit2 = (user_input // 1000) % 10
digit3 = (user_input // 100) % 10
digit4 = (user_input // 10) % 10
digit5 = user_input % 10

print("The individual digits are:")
# print(digit1, digit2, digit3, digit4, digit5, sep="\n")
print(digit1, digit2, digit3, digit4, digit5, sep="\n")
````

## Task 7

A driver has tracked several fuel fill-ups by recording the kilometers driven and the liters of gasoline used.
Write a script that prompts the user for the kilometers driven and the liters used for each fill-up. For each fill-up, compute and display the consumption in liters per kilometer (L/km).
The script should terminate using a sentinel value. When the user finishes, also display the overall average consumption (L/km) across all recorded fill-ups.

````python
# Initialize variables to track cumulative totals
total_kilometers = 0.0
total_liters = 0.0
fill_up_count = 0

print("Fuel Consumption Tracker (L/km)")
print("Enter -1 at the kilometers prompt to finish.\n")

while True:
    # Prompt the user for the kilometers driven
    kms_input = input(f"Enter kilometers driven for fill-up #{fill_up_count + 1} (or -1 to quit): ")
    
    # Check for the sentinel value to exit the loop
    if kms_input.strip() == "-1":
        break
        
    # Convert input to float for calculation
    kilometers = float(kms_input)
    
    # Prompt for liters used for this specific fill-up
    liters = float(input("Enter liters of gasoline used: "))
    
    # Compute and display consumption for the current fill-up
    if kilometers > 0:
        current_consumption = liters / kilometers
        print(f"-> Consumption for this fill-up: {current_consumption:.4f} L/km\n")
    else:
        print("-> Kilometers must be greater than 0 to compute consumption.\n")
        continue

    # Accumulate running totals
    total_kilometers += kilometers
    total_liters += liters
    fill_up_count += 1

# Summary calculation after loop terminates
print("\n--- Final Summary ---")
if fill_up_count > 0 and total_kilometers > 0:
    overall_average = total_liters / total_kilometers
    print(f"Total fill-ups recorded: {fill_up_count}")
    print(f"Total distance driven: {total_kilometers:.1f} km")
    print(f"Total fuel consumed: {total_liters:.1f} liters")
    print(f"Overall average consumption: {overall_average:.4f} L/km")
else:
    print("No valid fill-up data was entered.")
````

### Task 8 – Factorial
The factorial of a non-negative integer n is written as n! (read as “n factorial”) and is defined as:

```
n! = n · (n - 1) · (n - 2) · … · 1
```
By definition, the factorial of zero is 1 (that is, 0! = 1).

Write a script that reads a non-negative integer and computes its factorial.
Be aware that factorial values grow very quickly and can become very large.

````python
print("Factorial Calculator (n!)")

# Read a non-negative integer from the user with basic validation
while True:
    try:
        n = int(input("Enter a non-negative integer: "))
        if n >= 0:
            break
        print("Please enter a positive number or zero.")
    except ValueError:
        print("Invalid input. Please enter a valid integer.")

# Calculate the factorial
factorial = 1

# 0! and 1! will bypass the loop range and correctly output 1
for i in range(1, n + 1):
    factorial *= i

# Display the final calculated result
print(f"\nThe factorial of {n} ({n}!) is:")
print(factorial)
````
