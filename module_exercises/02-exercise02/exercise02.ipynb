## Task 1

Modify the following script to validate its input.
If the user enters any value other than 1 or 2, the input should be ignored and the user should be prompted to enter a new value.
Use a counter to keep track of the number of invalid entries.
When the program ends, display the number of invalid entries along with the existing output.

```
PASS = 1
FAIL = 2

passed = 0
failed = 0

for student in range(1, 11):  
    result = int(input(f'Enter the result for student {student} (1 = passed, 2 = failed): '))

    if result == PASS:
        passed += 1
    else:
        failed += 1
        
print(f'Passed: {passed}')
print(f'Failed: {failed}')
```

````python
# Accepted values from user
PASS = 1
FAIL = 2

# Counter variable to track how many times the user's invalid entries 
invalid = 0

# Counter to track passed and failed students
passed = 0
failed = 0

# Results for 10 students
for student in range(1,11):
    # For invalid user input, keep looping 
    while True:
        result = int(input(
            f"Enter the result for student: {student},"
            f"(1 = passed, 2 = failed): "
            )
        )
    
        # Validation and error handling function (check the value)
        if result == PASS:
            passed+=1
            break
        
        elif result == FAIL:
            failed+=1
            break
        
        else:
            # invalid number in counter 
            invalid+=1
            print("Invalid input, Please enter 1 or 2.")
    
# Results
print(f'Passed: {passed}')
print(f'Failed: {failed}')
print(f'Invalid: {invalid}')
````

## Task 2
Create a table for the numbers 0–5 that displays the number, its square, and its cube.

Use a for loop and f-strings to produce the following table, with the values in each column right-aligned:

```
number  square  cube
     0       0     0
     1       1     1
     2       4     8
     3       9    27

```

Hint: You can right-align the output like this: `print(f'{value:>6}')`

Here, `f'{value:>6}'` means that the value stored in the variable `value` is displayed in a field that is 6 characters wide, aligned to the right.

````python
# Create header
print(f"{"number":<6} {"square":>10} {"cube":>8}")
print("---"*9)


# Loop the numbers, get square and cube
for numbers in range(0,6): # 6
    # print(numbers)
    square = numbers **2
    cube = numbers**3

    print(f"{numbers:<6} {square:>8} {cube:>10}")
````

## Task 4
Write a script that reads a five-digit positive integer and then displays each digit on its own line.

For example, the number 54823 should be displayed as:

```
5
4
8
2
3
```
Derive the individual digits using the operators // and %.

````python
# Invalid user entries count
invalid = 0

# 
while True:
    user_input = input("Enter a five digit number: ")

    # Check the input is 5 digits long 
    if len(user_input) == 5 and user_input.isdigit():
        user_input = int(user_input)
        break

    # Handle invalid inputs
    else:
        invalid+=1
        print("Invalid inputs:")

# Derive the individual digits using the operators // and %.
digit1 = user_input // 10000
digit2 = (user_input // 1000) % 10
digit3 = (user_input // 100) % 10
digit4 = (user_input // 10) % 10
digit5 = user_input % 10

print("The individual digits are:")
# print(digit1, digit2, digit3, digit4, digit5, sep="\n")
print(digit1, digit2, digit3, digit4, digit5, sep="\n")
````
